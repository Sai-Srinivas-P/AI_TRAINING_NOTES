# Python Collections in Machine Learning, AI, and GenAI
Real-world examples of Lists, Dictionaries, Tuples, and Sets.

## 📋 Lists - Storing Ordered Data

```
# ML Example: Storing binary classification predictions
predictions = [0, 1, 0, 1, 1]
print("Model Predictions:", predictions)

# GenAI Example: List of prompts
prompts = ["Write a poem", "Generate a summary", "Translate this sentence"]
print("Batch Prompts for GenAI:", prompts)
```

## 📘 Dictionaries - Key-Value Pairs for Mapping

```
# ML Example: Model accuracy scores
model_scores = {
    "logistic_regression": 0.86,
    "random_forest": 0.89,
    "xgboost": 0.91
}
print("Model Scores:", model_scores)

# GenAI Example: Prompt structure for OpenAI API
prompt_template = {
    "role": "assistant",
    "content": "Please summarize the following text..."
}
print("Prompt Template:", prompt_template)
```

## 🧵 Tuples - Immutable and Ordered

```
# ML Example: Image input shape
input_shape = (28, 28)
print("Input shape for CNN:", input_shape)

# GenAI Example: (role, message) pairs
chat_log = ("user", "Generate a tweet about climate change.")
print("Chat Log Entry:", chat_log)
```

## 🧮 Sets - Unique and Unordered Values

```
# ML Example: Unique class labels
labels = set(["cat", "dog", "dog", "cat", "bird"])
print("Unique Labels:", labels)

# GenAI Example: Unique generations
generated_titles = {"AI in Healthcare", "AI in Finance", "AI in Healthcare"}
print("Unique Generated Titles:", generated_titles)
```

## ✅ Summary
| Type | Use Case | Real Example |
|------|-----------|--------------|
| List | Ordered Data | Predictions, Prompts |
| Dict | Key-Value Map | Model Scores, Prompt Templates |
| Tuple | Immutable Data | Input Shape, Chat Log |
| Set | Unique Values | Class Labels, Unique Outputs |
