```python
# Iris Clustering using KMeans and KMeans++

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

#StandardScaler - standardizes the features so that all columns have equal weight in clustering

# Load dataset
iris = load_iris()
X = iris.data

# Scale data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Regular K-Means (default init='random')
kmeans_random = KMeans(n_clusters=3, init='random', n_init=10, random_state=42)
kmeans_random.fit(X_scaled)
#K-means uses Euclidean distance, so larger numbers dominate calculations

# K-Means++ (smart initialization)
kmeans_plus = KMeans(n_clusters=3, init='k-means++', n_init=10, random_state=42)
kmeans_plus.fit(X_scaled)

#Random vs K-Means++
#Random - picks centers totally at random
#K-Means++ - pick the first center randomly then choses the next center far apart from each other (proper spread)

# Compare inertia (lower = better)
print("Random Init Inertia:", kmeans_random.inertia_)
print("K-Means++ Inertia:", kmeans_plus.inertia_)

#inertia - measures how tightly clusters are formed
#lower inertia = better clustering (points are closer to their centers)

# Plot comparison
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=kmeans_random.labels_, cmap='rainbow')
plt.title("Random Initialization")

plt.subplot(1,2,2)
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=kmeans_plus.labels_, cmap='rainbow')
plt.title("K-Means++ Initialization")
plt.show()
```

```python
# Mall Customer Segmentation using K-Means++

import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Load dataset (from Kaggle)
data = pd.read_csv("D:\\Phani\\etechguild\\syllabus\\demo\\lessons\\python\\Unsupervised Learning Clustering-18\\Mall_Customers.csv")
X = data[['Annual Income (k$)', 'Spending Score (1-100)']]

# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# K-Means++
kmeans_plus = KMeans(n_clusters=5, init='k-means++', n_init=10, random_state=42)
clusters = kmeans_plus.fit_predict(X_scaled)

# Add cluster column
data['Cluster'] = clusters

# Visualize
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=clusters, cmap='rainbow')
plt.title("Mall Customers - KMeans++ Clustering")
plt.xlabel("Annual Income (scaled)")
plt.ylabel("Spending Score (scaled)")
plt.show()
```
