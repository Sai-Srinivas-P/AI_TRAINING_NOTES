```python
#Student Performance Grouping
#Goal: Cluster students based on Math and Science marks
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Sample data
data = {'Math': [90, 85, 60, 58, 45, 82],
        'Science': [92, 88, 55, 62, 40, 80]}
df = pd.DataFrame(data)

# K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42,n_init='auto')
df['Cluster'] = kmeans.fit_predict(df)

# fit() - learns patterns from the data (finds cluster centers)
#predict() - assigns each student to the nearest cluster centre
# Visualize clusters
plt.scatter(df['Math'], df['Science'], c=df['Cluster'], cmap='viridis', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='Centroids')
plt.xlabel("Math Marks")
plt.ylabel("Science Marks")
plt.title("Student Performance Clusters")
plt.legend()
plt.show()

print(df)
```

```python
#Customer Segmentation (Retail)
#Goal: Segment customers by income and spending score
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Sample data
data = {'Income': [80000, 90000, 30000, 40000, 120000],
        'SpendingScore': [85, 88, 25, 40, 20]}
df = pd.DataFrame(data)

# K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42,n_init='auto')
df['Cluster'] = kmeans.fit_predict(df)

# Visualization
plt.scatter(df['Income'], df['SpendingScore'], c=df['Cluster'], cmap='Accent', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='Centroids')
plt.xlabel("Annual Income (₹)")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation")
plt.legend()
plt.show()

print(df)
```

```python
#Credit Card Fraud Detection (Outlier Identification)
#Goal: Detect transactions that don’t fit normal spending patterns
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Transaction data
data = {'Amount': [2500, 3000, 50000, 5000, 2800],
        'NumPurchases': [10, 12, 1, 20, 11]}
df = pd.DataFrame(data)

# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=42,n_init='auto')
df['Cluster'] = kmeans.fit_predict(df)

# Visualization
plt.scatter(df['Amount'], df['NumPurchases'], c=df['Cluster'], cmap='coolwarm', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='black', marker='X', s=200, label='Centroids')
plt.xlabel("Transaction Amount (₹)")
plt.ylabel("Number of Purchases")
plt.title("Credit Card Transaction Clusters")
plt.legend()
plt.show()

print(df)
```

```python
#E-Commerce Customer Segmentation
#Goal: Cluster customers by shopping frequency and order value
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# E-commerce customer data
data = {'OrdersPerMonth': [20, 3, 5, 15],
        'AvgOrderValue': [3000, 15000, 2000, 3500]}
df = pd.DataFrame(data)

# K-Means
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(df)

# Plot
plt.scatter(df['OrdersPerMonth'], df['AvgOrderValue'], c=df['Cluster'], cmap='plasma', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='black', marker='X', s=200, label='Centroids')
plt.xlabel("Orders per Month")
plt.ylabel("Avg Order Value (₹)")
plt.title("E-commerce Customer Segmentation")
plt.legend()
plt.show()

print(df)
```

```python
#Patient Health Profiling
#Goal: Group patients by BMI and Blood Pressure
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Patient data
data = {'BMI': [22, 24, 30, 32, 28],
        'BP': [120, 118, 150, 160, 130]}
df = pd.DataFrame(data)

# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=42)
df['Cluster'] = kmeans.fit_predict(df)

# Visualize
plt.scatter(df['BMI'], df['BP'], c=df['Cluster'], cmap='viridis', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='Centroids')
plt.xlabel("BMI")
plt.ylabel("Blood Pressure")
plt.title("Patient Health Clusters")
plt.legend()
plt.show()

print(df)
```

```python
#Real Estate Price Segmentation
#Goal: Cluster properties by price and area
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Real estate data
data = {'Area_sqft': [900, 1200, 2000, 1800],
        'Price_lakhs': [40, 55, 120, 110]}
df = pd.DataFrame(data)

# K-Means clustering
kmeans = KMeans(n_clusters=2, random_state=42)
df['Cluster'] = kmeans.fit_predict(df)

# Plot
plt.scatter(df['Area_sqft'], df['Price_lakhs'], c=df['Cluster'], cmap='Accent', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='black', marker='X', s=200, label='Centroids')
plt.xlabel("Area (sq ft)")
plt.ylabel("Price (₹ Lakhs)")
plt.title("Property Price Segmentation")
plt.legend()
plt.show()

print(df)
```

```python
#City Transport Optimization
#Goal: Cluster passenger pickup points to decide optimal bus stops
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Latitude & Longitude data
data = {'Latitude': [12.91, 12.93, 12.95, 13.02],
        'Longitude': [77.60, 77.62, 77.66, 77.71]}
df = pd.DataFrame(data)

# K-Means
kmeans = KMeans(n_clusters=2, random_state=42)
df['Cluster'] = kmeans.fit_predict(df)

# Visualization
plt.scatter(df['Latitude'], df['Longitude'], c=df['Cluster'], cmap='cool', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='Bus Stop (Centroid)')
plt.xlabel("Latitude")
plt.ylabel("Longitude")
plt.title("Optimal Bus Stop Clustering")
plt.legend()
plt.show()

print(df)
```
