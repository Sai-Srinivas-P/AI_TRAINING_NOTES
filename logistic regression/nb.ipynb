```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Simulate data
np.random.seed(42) # ensures reprodciblity
n = 200 #no.of students/data points
study_hours = np.random.normal(5, 2, n)
#study hours - normally distributed around 5 hours/week
attendance = np.random.normal(75, 10, n)
#attendance - normally distributed around 75%
pass_fail = (study_hours + attendance/20 + np.random.normal(0, 1, n)) > 10
#student pass if their combined score exceeds 10
#noise np.random.normal(0, 1, n) - add realistic randomness 
labels = np.where(pass_fail, 1, 0)

df = pd.DataFrame({'StudyHours': study_hours, 'Attendance': attendance, 'Pass': labels})

# Train-test split
X = df[['StudyHours', 'Attendance']]
y = df['Pass']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
nb_preds = nb.predict(X_test)

# Logistic Regression
lr = LogisticRegression()
lr.fit(X_train, y_train)
lr_preds = lr.predict(X_test)

# Accuracy comparison
print("Naive Bayes Accuracy:", accuracy_score(y_test, nb_preds))
print("Logistic Regression Accuracy:", accuracy_score(y_test, lr_preds))
```

```python
#Logistic regression is more robust to the added noise
#NB treats noise as part of feature distribution, which can distort itss probability estimates
#LR - doesn't care about independence - it models the boundary 
```
