```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Example dataset: BMI vs Diabetes (1=Yes, 0=No)
X = np.array([[18], [20], [22], [25], [27], [30], [32]])  # BMI values
y = np.array([0, 0, 0, 1, 1, 1, 1])  # labels (0=No Diabetes, 1=Diabetes)

# Train logistic regression model
model = LogisticRegression()
model.fit(X, y)

# Generate predictions (probabilities)
x_test = np.linspace(15, 35, 200).reshape(-1, 1)
y_prob = model.predict_proba(x_test)[:, 1]

# Find decision boundary (BMI where probability = 0.5)
decision_boundary = -model.intercept_[0] / model.coef_[0][0]

# Plot the sigmoid curve
plt.figure(figsize=(7,5))
plt.scatter(X, y, color='red', label="Actual data (0=No Diabetes, 1=Diabetes)")
plt.plot(x_test, y_prob, color='blue', label="Sigmoid Prediction Curve")
plt.axhline(0.5, color="gray", linestyle="--", label="Decision Threshold (0.5)")
plt.axvline(decision_boundary, color="green", linestyle="--", label=f"Decision Boundary ≈ {decision_boundary:.2f} BMI")

plt.title("Logistic Regression using Sigmoid Function (Diabetes Example)")
plt.xlabel("BMI")
plt.ylabel("Probability of Diabetes")
plt.legend(loc="best")
plt.grid(True)
plt.show()

# Print probabilities and classifications for specific BMIs
sample_bmi = np.array([[20], [25], [28], [32]])
pred_probs = model.predict_proba(sample_bmi)[:, 1]
pred_classes = model.predict(sample_bmi)

for b, p, c in zip(sample_bmi.flatten(), pred_probs, pred_classes):
    result = "Diabetes" if c == 1 else "No Diabetes"
    print(f"BMI = {b} --> Probability of Diabetes = {p:.2f} --> Prediction = {result}")
```
