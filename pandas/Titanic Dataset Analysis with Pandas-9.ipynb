```python
#Step 1: Load the Dataset
import pandas as pd
# Load Titanic dataset from a public GitHub mirror
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
print("Top 5 rows:\n", df.head())
print("\nBottom 5 rows:\n", df.tail())
```

```python
print("\nInfo:\n")
df.info()

print("\nDescribe:\n")
print(df.describe())
```

```python
print("\nShape:", df.shape)      # (rows, columns)
print("Size:", df.size)          # total number of elements
print("Number of Dimensions:", df.ndim)  # 2 for DataFrame
```

```python
print("\nData Types:\n", df.dtypes)
print("\nColumn Names:\n", df.columns)
print("\nIndex Info:\n", df.index)
```

```python
print("\nSurvival Counts:\n", df['Survived'].value_counts())
print("\nEmbarked Port Counts:\n", df['Embarked'].value_counts())
```

```python
# Check missing values
print("Missing values per column:\n", df.isnull().sum())

# Fill missing 'Age' with median
df['Age'].fillna(df['Age'].median(), inplace=True)

# Fill missing 'Embarked' with mode
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Drop 'Cabin' due to too many missing values
df.drop(columns=['Cabin'], inplace=True)

# Verify again
print("\nMissing values after handling:\n", df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set(style="whitegrid")

# Survival count
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()

# Age distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()

# Survival by gender
plt.figure(figsize=(6,4))
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Gender")
plt.show()

# Fare distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Fare'], bins=40, kde=True)
plt.title("Fare Distribution")
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Encode categorical variables
label_enc = LabelEncoder()
df['Sex'] = label_enc.fit_transform(df['Sex'])           # male=1, female=0
df['Embarked'] = label_enc.fit_transform(df['Embarked']) # C=0, Q=1, S=2

# Feature scaling for 'Age' and 'Fare'
scaler = StandardScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

# Drop non-numeric or irrelevant columns
df.drop(columns=['Name', 'Ticket', 'PassengerId'], inplace=True)

# Final check
print("\nPreprocessed DataFrame:\n", df.head())
```
