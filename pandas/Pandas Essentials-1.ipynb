```python
#Creating DataFrames
import pandas as pd

# Create a simple DataFrame
df_students = pd.DataFrame({
    'StudentID': [101, 102, 103],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [88, 92, 85],
    'City': ['Delhi', 'Mumbai', 'Delhi']
})
print("Created DataFrame:\n", df_students)
```

```python
#Importing CSV Files
# Load Titanic dataset from public GitHub mirror
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_titanic = pd.read_csv(url)
print("\nImported Titanic Data:\n", df_titanic.head())
```

```python
#Reading & Summarizing
print("\nInfo:\n")
df_titanic.info()

print("\nDescribe:\n")
print(df_titanic.describe())

print("\nShape:", df_titanic.shape)
print("Columns:", df_titanic.columns)
print("Index:", df_titanic.index)
```

```python
#Sorting & Indexing
# Sort by Fare
df_sorted = df_titanic.sort_values(by='Fare', ascending=False)
print("\nTop 5 Fares:\n", df_sorted[['Name', 'Fare']].head())

# Set index to PassengerId
df_indexed = df_titanic.set_index('PassengerId')
print("\nIndexed by PassengerId:\n", df_indexed.head())

# Reset index
df_reset = df_indexed.reset_index()
```

```python
#Merging DataFrames
# Create another DataFrame with student scores
df_scores = pd.DataFrame({
    'StudentID': [101, 103, 104],
    'AI_Score': [90, 85, 88]
})

# Merge with df_students
df_merged = pd.merge(df_students, df_scores, on='StudentID', how='inner')
print("\nMerged DataFrame:\n", df_merged)
```

```python
#Pivoting & Grouping
# Pivot: Max score per city
pivot_table = df_students.pivot_table(index='City', values='Score', aggfunc='max')
print("\nPivot Table (Max Score per City):\n", pivot_table)

# Grouping: Average score per city
grouped = df_students.groupby('City')['Score'].mean()
print("\nGrouped (Mean Score per City):\n", grouped)
```
