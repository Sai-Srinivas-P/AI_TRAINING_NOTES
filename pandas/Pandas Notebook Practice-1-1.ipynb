```python
#1. Setup & Sample Data Creation
import pandas as pd
import numpy as np

# Simulated student dataset with AI relevance
df = pd.DataFrame({
    'StudentID': [101, 102, 103, 104],
    'City': ['Mumbai', 'Delhi', 'Chennai', 'Mumbai'],
    'Temperature': [34.5, 31.0, 36.2, 33.8],
    'Math_Score': [88, 92, 78, 85],
    'AI_Score': [93, 87, 82, 90],
    'Pass': [True, True, False, True]
})

df.head()
```

```python
#2. Exploring the Data
# Shape and summary
print(df.shape)
print(df.info())
print(df.describe())

# Value counts for a binary column
print(df['Pass'].value_counts())
```

```python
#3. Filtering & Selection
# Students who scored above 85 in AI
df_high_ai = df[df['AI_Score'] > 85]

print(df.iloc[2])     # Row by position
```

```python
#4. Sorting & Ranking
# Sort by AI Score descending
df.sort_values(by='AI_Score', ascending=False, inplace=True)

# Rank students
df['Rank'] = df['AI_Score'].rank(ascending=False)
```

```python
#5. Aggregation & Grouping
# Mean scores by City
city_scores = df.groupby('City')[['Math_Score', 'AI_Score']].mean()

# Pivot table showing max scores by City
df.pivot_table(index='City', values=['Math_Score', 'AI_Score'], aggfunc='max')
```

```python
#6. Handling Missing Data
# Introduce some NaNs artificially
df.loc[1, 'AI_Score'] = np.nan

# Fill or drop
df.fillna(df['AI_Score'].mean(), inplace=True)
```

```python
#7. Combining DataFrames for ML Preprocessing
# Simulate test scores in another semester
df2 = pd.DataFrame({
    'StudentID': [101, 103, 104, 105],
    'AI_Score_2': [91, 88, 87, 77]
})

# Merge on StudentID
merged = pd.merge(df, df2, on='StudentID', how='inner')
```

```python
#8. Feature Engineering: ML Style
# Feature: score improvement
merged['AI_Improvement'] = merged['AI_Score_2'] - merged['AI_Score']

# Normalized Score
merged['AI_Normalized'] = (merged['AI_Score'] - merged['AI_Score'].mean()) / merged['AI_Score'].std()
```
