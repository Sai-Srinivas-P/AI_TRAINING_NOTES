```python
#Convert Strings to DateTime with to_datetime()
import pandas as pd

# Sample date strings
df = pd.DataFrame({
    'Event': ['Launch', 'Update', 'Review'],
    'Date': ['2025-08-01', '2025-08-05', '2025-08-10']
})

# Convert to datetime
df['Date'] = pd.to_datetime(df['Date'])
print("\nConverted to datetime:\n", df)
```

```python
# # Generate daily dates from Aug 1 to Aug 7
date_series = pd.date_range(start='2025-08-01', end='2025-08-07', freq='D')
print("\nGenerated Date Range:\n", date_series)
#freq='D', or 'B' or 'Q' or H
```

```python
# Extract year, month, day from Date column
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.day_name()
print("\nDate Components Extracted:\n", df)
```

```python
# Filter events in August
df_august = df[df['Date'].dt.month == 8]
print("\nEvents in August:\n", df_august)

# Filter events on Monday
df_monday = df[df['Date'].dt.day_name() == 'Monday']
print("\nEvents on Monday:\n", df_monday)
```
