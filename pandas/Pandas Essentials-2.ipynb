```python
#Sample DataFrame
import pandas as pd

# Sample employee data
df = pd.DataFrame({
    'EmpID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR'],
    'Salary': [50000, 60000, 55000, 62000, 52000],
    'Experience': [2, 5, 3, 6, 2]
})
```

```python
#Basic inspection
print("Head:\n", df.head())       # First 5 rows
print("\nTail:\n", df.tail())     # Last 5 rows
print("\nInfo:\n")
df.info()                         # Column types, non-null counts
print("\nDescribe:\n", df.describe())  # Summary stats for numeric columns
```

```python
#Label-based & Position-based Access
# loc: label-based
print("\nloc (Row 0, 'Name'):\n", df.loc[0, 'Name'])

# iloc: position-based
print("\niloc (Row 2, Column 3):\n", df.iloc[2, 3])
```

```python
#Sorting Values
# Sort by Salary descending
df_sorted = df.sort_values(by='Salary', ascending=False)
print("\nSorted by Salary:\n", df_sorted)
```

```python
#Labelling & Slicing
# Slice rows 1 to 3 and columns 'Name' to 'Salary'
df_slice = df.loc[1:3, 'Name':'Salary']
print("\nLabelled Slice:\n", df_slice)

# Slice using iloc: rows 0 to 2, columns 1 to 3
df_pos_slice = df.iloc[0:3, 1:4]
print("\nPositional Slice:\n", df_pos_slice)
```

```python
#Joins (Merge)
# Another DataFrame with bonus info
df_bonus = pd.DataFrame({
    'EmpID': [101, 103, 104],
    'Bonus': [5000, 4000, 6000]
})

# Inner join on EmpID
df_joined = pd.merge(df, df_bonus, on='EmpID', how='inner')
print("\nJoined DataFrame:\n", df_joined)
```

```python
#Aggregation
# Aggregate salary stats
agg_result = df.agg({
    'Salary': ['mean', 'sum', 'count'],
    'Experience': ['mean', 'sum']
})
print("\nAggregated Stats:\n", agg_result)
```

```python
#GroupBy + Aggregation
# Group by Department and aggregate
grouped = df.groupby('Department').agg({
    'Salary': ['mean', 'sum', 'count'],
    'Experience': 'mean'
})
print("\nGrouped Aggregation:\n", grouped)
```
