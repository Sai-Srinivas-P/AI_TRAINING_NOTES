```python
import pandas as pd
import numpy as np

# Load dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
# Handle missing values
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df.drop(columns=['Cabin'], inplace=True)
```

```python
# Encode categorical variables
df['Sex'] = df['Sex'].map({'male': 1, 'female': 0})
df['Embarked'] = df['Embarked'].map({'C': 0, 'Q': 1, 'S': 2})
```

```python
# Drop irrelevant columns
df.drop(columns=['Name', 'Ticket', 'PassengerId'], inplace=True)
```

```python
# loc[]: Access by label
print("\nloc[] - Passenger class and age of first 5 rows:\n", df.loc[0:4, ['Pclass', 'Age']])

# iloc[]: Access by position
print("\niloc[] - First 5 rows and first 3 columns:\n", df.iloc[0:5, 0:3])

# at[]: Access single value by label
print("\nat[] - Age of passenger at index 10:\n", df.at[10, 'Age'])

# iat[]: Access single value by position
print("\niat[] - Fare of passenger at row 10, column 5:\n", df.iat[10, 5])
```

```python
# Check for duplicates
print("\nNumber of duplicate rows:", df.duplicated().sum())

# Drop duplicates
df = df.drop_duplicates()

# Confirm removal
print("Remaining rows after dropping duplicates:", df.shape[0])
```

```python
# Filter passengers aged between 20 and 30
age_filtered = df[df['Age'].between(20, 30)]
print("\nPassengers aged between 20 and 30:\n", age_filtered[['Age', 'Fare', 'Survived']].head())

# Filter fares between 50 and 100
fare_filtered = df[df['Fare'].between(50, 100)]
print("\nPassengers with fare between 50 and 100:\n", fare_filtered[['Fare', 'Pclass', 'Survived']].head())
```

```python
#Summary
#Method	Purpose
#loc[]	Access rows/columns by label
#iloc[]	Access rows/columns by position
#at[]	Access single value by label
#iat[]	Access single value by position
#drop_duplicates()	Remove duplicate rows
#between()	Filter rows based on range condition
```
