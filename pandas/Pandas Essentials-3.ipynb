# Pandas Data Handling & Cleaning Toolkit

```python
#readjson
import pandas as pd

# Sample JSON data
json_data = '''
[
    {"EmpID": 101, "Name": "Alice", "Department": "HR", "Salary": 50000},
    {"EmpID": 102, "Name": "Bob", "Department": "IT", "Salary": 60000},
    {"EmpID": 103, "Name": "Charlie", "Department": "Finance", "Salary": null},
    {"EmpID": 104, "Name": "David", "Department": "IT", "Salary": 62000},
    {"EmpID": 105, "Name": "Eva", "Department": "HR", "Salary": null}
]
'''

# Read JSON string
df = pd.read_json(json_data)
print("Read JSON:\n", df)
```

```python
#Export to CSV
#note: change to your corresponding local path otherwise it will give error
df.to_csv("D:\\Phani\\2025\\trainings\\ai200725\\employees.csv", index=False)
```

```python
#value_counts
# Count frequency of departments
dept_counts = df['Department'].value_counts()
print("\nDepartment Counts:\n", dept_counts)
```

```python
#Filtering
# Filter rows where Department is either 'HR' or 'Finance'
df_filtered_isin = df[df['Department'].isin(['HR', 'Finance'])]
print("\nFiltered with isin:\n", df_filtered_isin)

# Filter rows where Salary is between 55000 and 63000
df_filtered_between = df[df['Salary'].between(55000, 63000)]
print("\nFiltered with between:\n", df_filtered_between)
```

```python
#Handling Missing Data
# Fill missing salaries with mean
mean_salary = df['Salary'].mean()
df_filled = df.fillna({'Salary': mean_salary})
print("\nFilled Missing Salaries:\n", df_filled)

# Drop rows with any missing values
df_dropped = df.dropna()
print("\nDropped Rows with NaNs:\n", df_dropped)

# try using df_filled.dropna()
```

```python
#Replace Values
# Replace department names
df_replaced = df.replace({'Department': {'HR': 'Human Resources', 'IT': 'Tech'}})
print("\nReplaced Department Names:\n", df_replaced)
```
