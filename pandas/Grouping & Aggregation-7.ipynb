```python
import pandas as pd

# Sample sales data
df = pd.DataFrame({
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'SalesPerson': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Helen'],
    'UnitsSold': [10, 15, 12, 20, 14, 18, 11, 17],
    'Revenue': [1000, 1500, 1200, 2000, 1400, 1800, 1100, 1700]
})
```

```python
# Total revenue per region
total_revenue = df.groupby('Region')['Revenue'].sum()
print("\nTotal Revenue per Region:\n", total_revenue)
```

```python
# Multiple stats per region
region_stats = df.groupby('Region').agg({
    'UnitsSold': ['mean', 'sum', 'count', 'min', 'max'],
    'Revenue': ['mean', 'sum', 'min', 'max']
})
print("\nRegion-wise Aggregated Stats:\n", region_stats)
```

```python
# Group by Region and SalesPerson
detailed_stats = df.groupby(['Region', 'SalesPerson']).agg({
    'UnitsSold': 'sum',
    'Revenue': 'sum'
})
print("\nDetailed Stats by Region & SalesPerson:\n", detailed_stats)
```

```python
# Clean column names after aggregation
region_clean = df.groupby('Region').agg(
    UnitsMean=('UnitsSold', 'mean'),
    UnitsTotal=('UnitsSold', 'sum'),
    RevenueTotal=('Revenue', 'sum')
)
print("\nClean Aggregated Output:\n", region_clean)
```
