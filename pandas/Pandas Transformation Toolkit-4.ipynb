```python
import pandas as pd

# Sample employee data
df = pd.DataFrame({
    'EmpID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'IT', 'Finance', 'IT'],
    'Salary': [50000, 60000, 55000, 62000],
    'Experience': [2, 5, 3, 6]
})
```

```python
# Apply a lambda to calculate bonus as 10% of salary
df['Bonus'] = df['Salary'].apply(lambda x: x * 0.10)
print("\nBonus with apply:\n", df[['Name', 'Salary', 'Bonus']])
```

```python
# Map department codes to full names
dept_map = {'HR': 'Human Resources', 'IT': 'Tech', 'Finance': 'Accounting'}
df['Department'] = df['Department'].map(dept_map)
print("\nMapped Department Names:\n", df[['Name', 'Department']])
```

```python
# Convert EmpID to string
df['EmpID'] = df['EmpID'].astype(str)
print("\nEmpID as string:\n", df.dtypes)
```

```python
# Rename columns
df_renamed = df.rename(columns={'Salary': 'BaseSalary', 'Experience': 'YearsExp'})
print("\nRenamed Columns:\n", df_renamed.head())
```

```python
# Sort by Years of Experience descending
df_sorted = df_renamed.sort_values(by='YearsExp', ascending=False)
print("\nSorted by Experience:\n", df_sorted[['Name', 'YearsExp']])
```
