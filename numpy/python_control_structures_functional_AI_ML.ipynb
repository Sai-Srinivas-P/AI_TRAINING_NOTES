# Python Control Structures and Functional Programming
This notebook covers control structures and functional programming tools with real-time AI/ML examples.

## If-Else Statements
**What:** Used to make decisions based on conditions.

**Why (AI/ML):** Useful for evaluating thresholds (e.g., checking if a prediction is above a cutoff).

```
accuracy = 0.87
if accuracy > 0.85:
    print('Model performance is acceptable')
else:
    print('Model needs improvement')
```

## For Loop
**What:** Iterates over a sequence.

**Why (AI/ML):** Used for batch processing of data, training multiple models, etc.

```
models = ['RandomForest', 'SVM', 'XGBoost']
for model in models:
    print(f'Training {model}')
```

## While Loop
**What:** Repeats while a condition is true.

**Why (AI/ML):** Can be used for iterative convergence (e.g., model improvement loops).

```
accuracy = 0.7
iteration = 0
while accuracy < 0.85:
    accuracy += 0.03
    iteration += 1
    print(f'Iteration {iteration}: Accuracy = {accuracy:.2f}')
```

## List Comprehensions
**What:** Concise way to create lists.

**Why (AI/ML):** Useful for feature transformation, cleaning, filtering, etc.

```
scores = [0.67, 0.89, 0.91, 0.56]
labels = ['Accept' if score > 0.85 else 'Reject' for score in scores]
print(labels)
```

## Dictionary Comprehensions
**What:** Concise way to create dictionaries.

**Why (AI/ML):** Useful for mapping IDs to results, organizing predictions.

```
ids = [101, 102, 103]
preds = [0.8, 0.9, 0.95]
pred_dict = {id: pred for id, pred in zip(ids, preds)}
print(pred_dict)
```

## Functions
**What:** Reusable blocks of code.

**Why (AI/ML):** Common practice to define preprocessing or scoring steps as functions.

```
def scale_data(data, factor):
    return [x * factor for x in data]

print(scale_data([1, 2, 3], 0.1))
```

## Map
**What:** Applies a function to each item.

**Why (AI/ML):** Can apply transformations to features, predictions, etc.

```
data = [100, 200, 300]
scaled = list(map(lambda x: x / 10, data))
print(scaled)
```

## Filter
**What:** Filters items based on condition.

**Why (AI/ML):** Useful for selecting relevant data (e.g., confident predictions).

```
predictions = [0.65, 0.9, 0.45, 0.88]
confident = list(filter(lambda x: x > 0.8, predictions))
print(confident)
```

## Reduce
**What:** Applies rolling computation to elements.

**Why (AI/ML):** Used for aggregation like total score, weighted sums, etc.

```
from functools import reduce
values = [1, 2, 3, 4]
sum_values = reduce(lambda x, y: x + y, values)
print(sum_values)
```
