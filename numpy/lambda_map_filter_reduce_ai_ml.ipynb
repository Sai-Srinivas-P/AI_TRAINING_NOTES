# Lambda, Map, Filter, and Reduce in Python

This notebook demonstrates the usage of anonymous functions (lambda) and how they are used with `map`, `filter`, and `reduce` functions. These are frequently used in data processing workflows, especially in Machine Learning and AI for data transformations and feature engineering.

```python
# Traditional function to square a number
def square(x):
    return x * x

# Lambda version
square_lambda = lambda x: x * x

print("Traditional:", square(5))
print("Lambda:", square_lambda(5))
```

## Using Lambda with `map()`
`map()` applies a function to every item in an iterable. Useful in data cleaning, transformation.

```python
nums = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, nums))
print("Squared:", squared)
```

## Using Lambda with `filter()`
`filter()` keeps only items for which the function returns True. Useful in data filtering and cleaning.

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8]
even = list(filter(lambda x: x % 2 == 0, nums))
print("Even Numbers:", even)
```

## Using Lambda with `reduce()`
`reduce()` applies a rolling computation to sequential pairs. Useful in aggregation tasks.

**Note**: Requires `functools` module.

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, nums)
print("Product:", product)
```

## Real-World Example: Sorting ML Models by Accuracy

Sorting list of model tuples using lambda to extract accuracy.

```python
models = [("SVM", 0.88), ("RandomForest", 0.93), ("LogReg", 0.85)]
sorted_models = sorted(models, key=lambda x: x[1], reverse=True)
print("Sorted Models:", sorted_models)
```
