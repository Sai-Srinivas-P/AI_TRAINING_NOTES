# map(), filter(), reduce() in Python for AI/ML

This notebook explains how `map`, `filter`, and `reduce` functions work in Python, and how they are used in AI/ML pipelines.

```python
# map() example - Apply transformation to each element
nums = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, nums))
print("Squared:", squared)  # [1, 4, 9, 16, 25]
```

**AI/ML Use Case:** Apply transformation to features, e.g., normalization or converting text to lowercase.

```python
sentences = ["Hello World", "Machine Learning"]
lower = list(map(lambda s: s.lower(), sentences))
print(lower)  # ['hello world', 'machine learning']
```

---
# filter() example - Keep only values meeting a condition

```python
nums = [1, 2, 3, 4, 5]
even = list(filter(lambda x: x % 2 == 0, nums))
print("Even numbers:", even)  # [2, 4]
```

**AI/ML Use Case:** Clean dataset by removing empty strings or filtering invalid values.

```python
tokens = ["the", "", "data", "", "science"]
non_empty = list(filter(lambda x: x != "", tokens))
print(non_empty)  # ['the', 'data', 'science']
```

---
# reduce() example - Aggregate all values into one
Requires `functools` module.

```python
from functools import reduce
nums = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, nums)
print("Product:", product)  # 120
```

**AI/ML Use Case:** Aggregate strings or compute total score/accuracy.

```python
words = ["machine", "learning", "rocks"]
sentence = reduce(lambda x, y: x + " " + y, words)
print(sentence)  # 'machine learning rocks'
```

| Function  | Purpose                          | Output         | Use Case in AI/ML                            |
|-----------|----------------------------------|----------------|----------------------------------------------|
| `map()`   | Apply transformation             | Transformed list | Feature scaling, text transformation        |
| `filter()`| Select based on condition        | Filtered list   | Data cleaning, outlier removal              |
