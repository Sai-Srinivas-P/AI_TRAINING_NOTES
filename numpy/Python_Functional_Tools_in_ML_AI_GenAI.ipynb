# Python Essentials in ML, AI, and GenAI
### Comprehensions, Functions, Map, Reduce, and Filter

## ✅ List Comprehensions – Efficient Data Processing

```
# ML Example: Convert prediction probabilities to binary
probs = [0.1, 0.7, 0.4, 0.9]
preds = [1 if p > 0.5 else 0 for p in probs]
print("Binary Predictions:", preds)

# GenAI Example: Format prompts
raw_prompts = ["translate", "summarize", "generate story"]
formatted = [f"Please {task} the following text:" for task in raw_prompts]
print("Formatted Prompts:", formatted)
```

## 🧠 Functions – Modular and Reusable Code

```
# ML Example: Accuracy function
def accuracy(y_true, y_pred):
    return sum([1 for a, b in zip(y_true, y_pred) if a == b]) / len(y_true)

y_true = [1, 0, 1, 1]
y_pred = [1, 0, 0, 1]
print("Accuracy:", accuracy(y_true, y_pred))

# GenAI Example: Prompt creator
def create_prompt(task, text):
    return f"Please {task} the following: {text}"

print(create_prompt("summarize", "AI is transforming every industry."))
```

## 🔁 Map – Apply a Function to a Collection

```
# ML Example: Round probabilities
from math import floor
probabilities = [0.83, 0.12, 0.55]
floored = list(map(floor, probabilities))
print("Floored Probabilities:", floored)

# GenAI Example: Capitalize responses
responses = ["hello there", "how can I help you?"]
capitalized = list(map(str.title, responses))
print("Capitalized:", capitalized)
```

## 🧹 Filter – Select Based on Condition

```
# ML Example: Filter high confidence predictions
preds = [0.2, 0.9, 0.65, 0.3]
high_conf = list(filter(lambda x: x > 0.6, preds))
print("High Confidence:", high_conf)

# GenAI Example: Filter short responses
answers = ["yes", "certainly", "no", "absolutely"]
long_only = list(filter(lambda s: len(s) > 3, answers))
print("Filtered Answers:", long_only)
```

## ➖ Reduce – Aggregate a Collection into a Single Value

```
# ML Example: Sum of squared errors
from functools import reduce
errors = [2, 3, 1]
squared_sum = reduce(lambda a, b: a + b**2, errors, 0)
print("Sum of Squared Errors:", squared_sum)

# GenAI Example: Combine all chatbot messages into one
messages = ["Hi", "How can I help?", "Here’s your answer."]
conversation = reduce(lambda a, b: a + " " + b, messages)
print("Conversation:", conversation)
```
