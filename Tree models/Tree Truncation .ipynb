```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

X, y = load_iris(return_X_y=True)

# A truncated tree
dt = DecisionTreeClassifier(
    criterion="gini", 
    max_depth=3, 
    min_samples_split=5, 
    min_samples_leaf=2, 
    max_features="sqrt",
    random_state=42
)

dt.fit(X, y)
print("Tree trained with truncation (pre-pruning).")
```

```python
#In this example:

#Tree depth is limited to 3.

#At least 5 samples are required to split.

#Each leaf must have at least 2 samples.

#At each split, only √features are considered.

#This forces the tree to be simpler and more generalizable.
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load dataset
X, y = load_iris(return_X_y=True)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
#Train decision trees with different truncation depths
depths = [2, 3, 5, None]  # None = unlimited depth
results = []

for d in depths:
    dt = DecisionTreeClassifier(max_depth=d, random_state=42)
    dt.fit(X_train, y_train)

    train_acc = accuracy_score(y_train, dt.predict(X_train))
    test_acc = accuracy_score(y_test, dt.predict(X_test))

    results.append((d, train_acc, test_acc))

# Display results
for r in results:
    print(f"Max Depth={r[0]} | Train Acc={r[1]:.2f} | Test Acc={r[2]:.2f}")
```

```python
#Shallow tree (max_depth=2):

    #Slightly lower training accuracy.

    #Good generalization → not overfitting.

#Moderate depth (max_depth=3):

     #High training accuracy.

     #Also high test accuracy → best balance.

#Deep tree (max_depth=5):

    #Perfect fit on training data.

    #Test accuracy starts dropping → signs of overfitting.

#Unlimited depth (None):

    #Tree memorizes training data (100% accuracy).

    #Poor test accuracy → clear overfitting.

#Key Learning: Controlling tree depth helps balance bias-variance trade-off.
```
