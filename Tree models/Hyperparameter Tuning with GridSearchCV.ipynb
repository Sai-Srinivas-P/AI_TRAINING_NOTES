```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load dataset
X, y = load_iris(return_X_y=True)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
#Define hyperparameter grid & apply GridSearchCV that uses cross - validation
```

```python
# Define parameter grid
param_grid = {
    "criterion": ["gini", "entropy"],
    "max_depth": [2, 3, 4, 5, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
}

# Initialize GridSearchCV
grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,                # 5-fold cross-validation
    scoring="accuracy",
    n_jobs=-1
)

# Fit the grid search
grid_search.fit(X_train, y_train)

#cv=5 , perform 5 fold cross validation that is split train data into 5 parts that is  trains on 4 parts and validates on 1. Repeat 5 times
# n_jobs=-1, use all available CPU cores for parallel processing - faster search
```

```python
#Check best parameters and accuracy
```

```python
print("Best Hyperparameters:", grid_search.best_params_)
print("Best Cross-Validation Score:", grid_search.best_score_)

# Evaluate on test set
best_model = grid_search.best_estimator_
test_acc = accuracy_score(y_test, best_model.predict(X_test))
print("Test Accuracy:", test_acc)
#Best Cross-Validation Score - average accuracy across 5-folds that model generalizes and robust
#test accuracy is 1 - it mean model correclty 100% of the test samples
```

```python

#GridSearchCV picked the best hyperparameters automatically.

#Notice that the “best” model balances training accuracy and validation accuracy.

#This prevents overfitting (deep trees memorizing) and underfitting (shallow trees missing patterns).
```
