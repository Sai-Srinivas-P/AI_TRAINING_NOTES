```python
# Step 1: Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

# Step 2: Load sample dataset (Iris flower classification)
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target

df = pd.DataFrame(iris.data, columns=iris.feature_names)
print(df.head())

print(iris.feature_names)
print(iris.target)
# Step 3: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 4: Define Hyperparameters and train Decision Tree
# Hyperparameters are set BEFORE training
dt = DecisionTreeClassifier(max_depth=3, criterion="gini", random_state=42)
dt.fit(X_train, y_train)

# Step 5: Check Accuracy
print("Accuracy on test set:", dt.score(X_test, y_test))

# Step 6: Parameters (learned from data)
print("\n=== Learned Parameters (Tree Structure) ===")
print("Number of features used:", dt.n_features_in_)
print("Number of classes:", dt.n_classes_)
print("Classes:", dt.classes_)
print("Tree depth (learned):", dt.get_depth())
print("Number of leaves (learned):", dt.get_n_leaves())

# Step 7: Visualize Decision Tree
plt.figure(figsize=(12,8))
plot_tree(dt, feature_names=iris.feature_names, 
          class_names=iris.target_names,
          filled=True, rounded=True)
plt.show()
```
