```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
# -----------------------------
# 1. Create a simple dataset
# -----------------------------
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Relationship": ["Single", "Married", "In a relationship", "Single", "Married"],
    "Age": [25, 32, 29, 40, 36],
    "Salary": [40000, 52000, 48000, 61000, 58000]
}
```

```python
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# -----------------------------
# 2. Handling Categorical Variable (Dummy Encoding)
# -----------------------------
# Create dummy variables for 'Relationship' and drop the first level to avoid dummy variable trap
df_encoded = pd.get_dummies(df, columns=["Relationship"], drop_first=True)

print("\nAfter Dummy Encoding (drop_first=True):")
print(df_encoded)
```

```python
# -----------------------------
# 3. Standardization (Z-score Scaling)
# -----------------------------
scaler = StandardScaler()
df_encoded[["Age_std", "Salary_std"]] = scaler.fit_transform(df_encoded[["Age", "Salary"]])

print("\nAfter Standardization:")
print(df_encoded[["Age", "Salary", "Age_std", "Salary_std"]])
```

```python
# -----------------------------
# 4. Min-Max Scaling
# -----------------------------
minmax = MinMaxScaler()
df_encoded[["Age_minmax", "Salary_minmax"]] = minmax.fit_transform(df_encoded[["Age", "Salary"]])

print("\nAfter Min-Max Scaling:")
print(df_encoded[["Age", "Salary", "Age_minmax", "Salary_minmax"]])
```
