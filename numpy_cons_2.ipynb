```python
## Creating NumPy Arrays

import numpy as np

# From a list
a = np.array([1, 2, 3])
print("1D Array:", a)

# 2D Array
b = np.array([[1, 2], [3, 4]])
print("2D Array:", b)

# Zeros and Ones
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
print("Zeros:", zeros)
print("Ones:", ones)
```

```python
## Array Indexing and Slicing

arr = np.array([10, 20, 30, 40, 50])
print("Original:", arr)
print("Element at index 1:", arr[1])
print("Slicing [1:4]:", arr[1:4])
```

```python
## Array Operations

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("Addition:", x + y)
print("Multiplication:", x * y)
print("Dot Product:", np.dot(x, y))
```

```python
## Reshaping and Broadcasting
import numpy as np

a = np.array([[1, 2], [3, 4], [5, 6]])
print("Original Shape:", a.shape)

reshaped = a.reshape((2, 3))
print("Reshaped:", reshaped)

# Broadcasting example
x = np.array([[1], [2], [3]])
y = np.array([10, 20, 30])
print("Broadcasted Sum:", x + y)
```

```python
## Visualizing NumPy Array as Image

import matplotlib.pyplot as plt

# RGB image using NumPy
rgb_array = np.array([
    [[255, 0, 0], [0, 255, 0], [0, 0, 255]],
    [[255, 255, 0], [0, 255, 255], [255, 0, 255]],
    [[255, 255, 255], [128, 128, 128], [0, 0, 0]]
], dtype=np.uint8)

plt.imshow(rgb_array)
plt.title("RGB Image using NumPy")
plt.axis('off')
plt.show()
```

```python
#Data Preparation (Vectors, Matrices, Normalization)

import numpy as np

# Dataset with features
data = np.array([[150, 60], [160, 65], [170, 72], [180, 80]])

# Normalizing (min-max scaling)
data_min = data.min(axis=0)
data_max = data.max(axis=0)
normalized = (data - data_min) / (data_max - data_min)

print("Normalized Data:", normalized)
```

```python
#One-Hot Encoding (for Classification Models)
import numpy as np
labels = np.array([0, 1, 2, 1])
num_classes = 3

# One-hot encode manually
one_hot = np.eye(num_classes)[labels]
print("One-hot Encoded: ", one_hot)
```

```python
#Vectorization (Faster than Loops in ML Training)

# Traditional loop (slow)
x = np.arange(1, 10001)
y = np.arange(1, 10001)
dot = 0
for i in range(len(x)):
    dot += x[i] * y[i]
print("Dot product using loop:", dot)

# Vectorized
dot_vec = np.dot(x, y)
print("Dot product using NumPy:", dot_vec)
```

```python
#Cost Function (MSE Example in Regression)

# Predictions vs Actual
y_true = np.array([3.0, -0.5, 2.0, 7.0])
y_pred = np.array([2.5, 0.0, 2.0, 8.0])

# y_true - y_pred = 3.0 - 2.5 = 0.5 (error), -ve or +ve = 
mse = np.mean((y_true - y_pred)**2)
print("Mean Squared Error:", mse)
#ML - y_true - y_pred - this captures how far off predictions are 
# if more MSE - not a better accuracy
```

```python
#Linear Regression (Weights Calculation with Normal Equation)

# Features and targets
X = np.array([[1, 1], [1, 2], [1, 3]])
y = np.array([1, 2, 3])

# Normal Equation: w = (X^T X)^-1 X^T y
w = np.linalg.inv(X.T @ X) @ X.T @ y
print("Weights:", w)
```

```python
#Broadcasting in Batch Predictions (Neural Networks)

# Inputs: 3 samples, 2 features
X = np.array([[0.1, 0.2], [0.4, 0.6], [0.7, 0.9]])
weights = np.array([0.5, 0.8])  # shape: (2,)

# Broadcasting dot product for all samples
predictions = X @ weights
print("Predictions:", predictions)

# A @ B - multiply matrix A with matrix B
```
