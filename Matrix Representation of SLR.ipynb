```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Data (same as our manual example)
X = np.array([[1, 1],   # column of 1’s for intercept + predictor values
              [1, 2],
              [1, 3]])

Y = np.array([[2],
              [3],
              [5]])

# Step 2: Apply Normal Equation -> β = (X^T X)^(-1) X^T Y
XT = X.T
beta = np.linalg.inv(XT @ X) @ XT @ Y

print("Beta coefficients (Intercept, Slope):")
print(beta)

# Step 3: Show regression line
intercept, slope = beta.flatten()
print(f"\nRegression Line: y = {intercept:.2f} + {slope:.2f}x")

# Step 4: Make predictions
Y_pred = X @ beta
print("\nPredicted Y values:")
print(Y_pred.flatten())
```

```python
# Step 5: Visualization
# Extract x-values (second column of X)
x_vals = X[:, 1]
y_vals = Y.flatten()

plt.scatter(x_vals, y_vals, color="blue", label="Actual Data")
plt.plot(x_vals, Y_pred, color="red", label="Regression Line")

plt.xlabel("X (Predictor)")
plt.ylabel("Y (Target)")
plt.title("Linear Regression using Normal Equation")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Step 5: Visualization - Regression Line
x_vals = X[:, 1]
y_vals = Y.flatten()

plt.figure(figsize=(12, 5))

# Plot 1: Regression line with data
plt.subplot(1, 2, 1)
plt.scatter(x_vals, y_vals, color="blue", label="Actual Data")
plt.plot(x_vals, Y_pred, color="red", label="Regression Line")
plt.xlabel("X (Predictor)")
plt.ylabel("Y (Target)")
plt.title("Linear Regression Fit")
plt.legend()
plt.grid(True)

# Plot 2: Residual plot
residuals = y_vals - Y_pred.flatten()
plt.subplot(1, 2, 2)
plt.scatter(x_vals, residuals, color="purple")
plt.axhline(y=0, color="black", linestyle="--")  # baseline
plt.xlabel("X (Predictor)")
plt.ylabel("Residuals (Y - Y_pred)")
plt.title("Residual Plot")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Step 5: Residuals
y_vals = Y.flatten()
y_pred_vals = Y_pred.flatten()
residuals = y_vals - y_pred_vals

# Step 6: R² Calculation
ss_res = np.sum(residuals ** 2)
ss_tot = np.sum((y_vals - np.mean(y_vals)) ** 2)
r2 = 1 - (ss_res / ss_tot)

print(f"\nR² (Coefficient of Determination): {r2:.4f}")
```

```python
# Step 7: Error metrics
mse = np.mean(residuals ** 2)
rmse = np.sqrt(mse)

print(f"\nR² (Coefficient of Determination): {r2:.4f}")
print(f"MSE (Mean Squared Error): {mse:.4f}")
print(f"RMSE (Root Mean Squared Error): {rmse:.4f}")
```

```python
# Step 8: Visualization
x_vals = X[:, 1]

plt.figure(figsize=(12, 5))

# Plot 1: Regression line with data
plt.subplot(1, 2, 1)
plt.scatter(x_vals, y_vals, color="blue", label="Actual Data")
plt.plot(x_vals, y_pred_vals, color="red", label="Regression Line")
plt.xlabel("X (Predictor)")
plt.ylabel("Y (Target)")
plt.title(f"Linear Regression Fit\nR²={r2:.2f}, RMSE={rmse:.2f}")
plt.legend()
plt.grid(True)

# Plot 2: Residual plot
plt.subplot(1, 2, 2)
plt.scatter(x_vals, residuals, color="purple")
plt.axhline(y=0, color="black", linestyle="--")  # baseline
plt.xlabel("X (Predictor)")
plt.ylabel("Residuals (Y - Y_pred)")
plt.title("Residual Plot")
plt.grid(True)

plt.tight_layout()
plt.show()
```
