```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Ad Spend
y = np.array([4, 6.5, 7, 9.5, 12])           # Revenue

#scikit-learn expects input features(X) to be in 2D -- (n_samples,n_features)
#scikit-learn expects target variable(y) to be in 1D
#for each sample of x - represents target variable in y
```

```python
# Create and train model
model = LinearRegression()
model.fit(X, y)   #internally computes best fit line by solving y=Bo+B1x

# Coefficients
print(f"Intercept (β₀): {model.intercept_:.2f}")
print(f"Slope (β₁): {model.coef_[0]:.2f}")

```

```python
# Predict revenue for ₹6 lakhs ad spend
predicted = model.predict([[6]])
print(f"Predicted Revenue for ₹6L ad spend: {predicted[0]:.2f} lakhs")

# Plotting
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, model.predict(X), color='red', label='Best-Fit Line')
plt.xlabel("Ad Spend (₹ lakhs)")
plt.ylabel("Revenue (₹ lakhs)")
plt.title("Simple Linear Regression: Ad Spend vs Revenue")
plt.legend()
plt.grid(True)
plt.show()
```
