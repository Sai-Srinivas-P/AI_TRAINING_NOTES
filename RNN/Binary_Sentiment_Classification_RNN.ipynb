```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, GRU, Bidirectional, Dense

# 1. Load dataset
vocab_size = 10000
maxlen = 200

print("Downloading IMDB dataset...")
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

# pad sequences
X_train = pad_sequences(X_train, maxlen=maxlen)
X_test = pad_sequences(X_test, maxlen=maxlen)

embedding_dim = 32

def build_and_train(model, title):
    print(f"\n=========== Training: {title} ===========")

    model.build(input_shape=(None, maxlen))  # build the model
    model.summary()                          # summary shows correct shapes & params

    model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])
    model.fit(X_train, y_train, epochs=2, batch_size=128, validation_data=(X_test, y_test))
    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    print(f"{title} → Test Accuracy: {acc:.4f}")

# =================== Vanilla RNN ===================
vanilla_rnn = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=maxlen),
    SimpleRNN(32),
    Dense(1, activation="sigmoid")
])
build_and_train(vanilla_rnn, "Vanilla RNN")

# ===================== LSTM ========================
lstm_model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=maxlen),
    LSTM(64),
    Dense(1, activation="sigmoid")
])
build_and_train(lstm_model, "LSTM")

# ====================== GRU ========================
gru_model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=maxlen),
    GRU(64),
    Dense(1, activation="sigmoid")
])
build_and_train(gru_model, "GRU")

# ============= Bidirectional LSTM ==================
bilstm_model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=maxlen),
    Bidirectional(LSTM(64)),
    Dense(1, activation="sigmoid")
])
build_and_train(bilstm_model, "Bidirectional LSTM")
```
