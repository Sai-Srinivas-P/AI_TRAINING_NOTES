# 🐍 Python Basics for AI/ML - Real Time Examples

## ✅ Control Structures

```python
# If-Else example: Health check
bp = 140
if bp > 130:
    print("⚠️ High Blood Pressure")
else:
    print("✅ Normal")
```

```python
# For loop: Predict multiple risk factors
factors = ["Diabetes", "Hypertension", "Obesity"]
for f in factors:
    print("Checking risk factor:", f)
```

```python
# While loop: Model retraining threshold
accuracy = 0.85
attempts = 0
while accuracy < 0.90:
    accuracy += 0.01
    attempts += 1
print("Model reached required accuracy in", attempts, "iterations")
```

```python
# List comprehension: Square all features
features = [1, 2, 3, 4, 5]
squared = [x**2 for x in features]
print(squared)
```

```python
# Dictionary comprehension: Label mapping
labels = ["cat", "dog", "mouse"]
label_dict = {label: idx for idx, label in enumerate(labels)}
print(label_dict)
```

## ✅ Functions

```python
def normalize(score):
    return score / 100

print(normalize(85))
```

## ✅ Lambda Functions + Map

```python
nums = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, nums))
print("Squared:", squared)
```

## ✅ Filter Function

```python
# Filter patients over age 60
ages = [45, 67, 34, 80, 25]
senior_citizens = list(filter(lambda x: x > 60, ages))
print(senior_citizens)
```

## ✅ Reduce Function

```python
from functools import reduce
total = reduce(lambda x, y: x + y, [10, 20, 30])
print("Sum:", total)
```

---
## 🎯 Real-time AI/ML Scenarios Explained
- **If/else**: Classification thresholds (e.g., fraud detection)
- **Loops**: Hyperparameter tuning, retraining
- **List/Dict comprehension**: Data transformation, preprocessing
- **Lambda/Map**: Vectorized feature transformations
- **Filter**: Cleaning outliers or selecting samples
- **Reduce**: Aggregating results (e.g., total loss)
