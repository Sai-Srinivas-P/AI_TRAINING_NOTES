```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("D:\\Phani\\etechguild\\syllabus\\demo\\lessons\\python\\EDA\\data sheets\\nas.csv")
```

```python
# Clean numeric columns
score_cols = ["Maths..", "Reading..", "Science..", "Social.."]
for col in score_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Clean age column
df["Age"] = df["Age"].str.replace(" years", "", regex=False)
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
# ---- Comparison of Averages ----

# 1. Overall averages
print("Overall Averages:\n", df[score_cols].mean())
```

```python
# 2. Gender comparison
print("\nAverage Scores by Gender:\n", df.groupby("Gender")[score_cols].mean())
```

```python
# 3. Poverty status
print("\nAverage Scores by Poverty:\n", df.groupby("Below.poverty")[score_cols].mean())
```

```python
# 4. Father's education
print("\nAverage Scores by Father's Education:\n", df.groupby("Father.edu")[score_cols].mean())
```

```python
# 5. Number of siblings
print("\nAverage Scores by Siblings:\n", df.groupby("Siblings")[score_cols].mean())
```

```python
# ---- Visualization ----
plt.figure(figsize=(8,6))
sns.barplot(x="Gender", y="Maths..", data=df)
plt.title("Average Maths Score by Gender")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x="Below.poverty", y="Reading..", data=df)
plt.title("Reading Score Distribution by Poverty Status")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
ax = sns.boxplot(x="Below.poverty", y="Reading..", data=df)

# Loop through each group to annotate stats
groups = df["Below.poverty"].unique()
for i, group in enumerate(groups):
    group_data = df[df["Below.poverty"] == group]["Reading.."].dropna()
    median = group_data.median()
    min_val = group_data.min()
    max_val = group_data.max()

    # Annotate median
    ax.text(i, median, f"Median: {median:.1f}", color="white", ha="center", va="bottom", fontsize=9)
    # Annotate min
    ax.text(i, min_val, f"Min: {min_val:.1f}", color="green", ha="center", va="bottom", fontsize=9)
    # Annotate max
    ax.text(i, max_val, f"Max: {max_val:.1f}", color="red", ha="center", va="top", fontsize=9)

plt.title("Reading Score Distribution by Poverty Status")
plt.ylabel("Reading Score")
plt.xlabel("Poverty Status")
plt.tight_layout()
plt.show()
```
