```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -----------------------------
# 1. Unordered Categorical Variables
# -----------------------------

# Example: Type of loan (no inherent order)
data_unordered = pd.DataFrame({
    'LoanType': ['Home', 'Personal', 'Auto', 'Personal', 'Home', 'Auto', 'Personal', 'Home']
})
```

```python
print("Unordered Categorical Variable Example:")
print(data_unordered['LoanType'].value_counts())
```

```python
# Visualization
plt.figure(figsize=(5,3))
data_unordered['LoanType'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Loan Types (Unordered Categorical)")
plt.ylabel("Count")
plt.show()
```

```python
# -----------------------------
# 2. Ordered Categorical Variables
# -----------------------------

# Example: Salary levels (Low < Medium < High)
salary_levels = ['Low', 'Medium', 'High']
data_ordered = pd.DataFrame({
    'Salary': pd.Categorical(
        ['High', 'Low', 'Medium', 'Medium', 'High', 'Low'],
        categories=salary_levels,
        ordered=True
    )
})
```

```python
print("\nOrdered Categorical Variable Example:")
print(data_ordered['Salary'].value_counts().sort_index())
```

```python
# Visualization
plt.figure(figsize=(5,3))
data_ordered['Salary'].value_counts().sort_index().plot(kind='bar', color='orange')
plt.title("Salary Levels (Ordered Categorical)")
plt.ylabel("Count")
plt.show()
```

```python
# -----------------------------
# 3. Quantitative / Numeric Variables
# -----------------------------

# Example: Age of people
data_numeric = pd.DataFrame({
    'Age': [22, 25, 47, 35, 52, 46, 33, 29, 41, 60]
})
```

```python
print("\nQuantitative / Numeric Variable Example:")
print(data_numeric.describe())
```

```python
# Visualization
plt.figure(figsize=(5,3))
plt.hist(data_numeric['Age'], bins=5, color='green', edgecolor='black')
plt.title("Age Distribution (Quantitative Variable)")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
