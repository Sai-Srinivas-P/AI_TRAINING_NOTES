```python
#univariate analysis of the shares column from the News Popularity dataset.
#This will help us understand how article popularity varies across:

# Days of the week

# Weekends vs weekdays

# Article types (e.g., tech, lifestyle, entertainment)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_excel('D:\\Phani\\etechguild\\syllabus\\demo\\lessons\\python\\EDA\\data sheets\\EDA_news_popularity.xlsx')
```

```python
# -------------------------------
# Summary Metrics for 'shares'
# -------------------------------
print("Summary Statistics of Shares:")
print(df['shares'].describe())
```

```python
#Boxplot: Shares across weekdays
weekdays = [" weekday_is_monday", " weekday_is_tuesday", " weekday_is_wednesday",
            " weekday_is_thursday", " weekday_is_friday", " weekday_is_saturday", " weekday_is_sunday"]

# Create a new column for weekday name
for i, day in enumerate(weekdays, start=1):
    df.loc[df[day] == 1, 'Day'] = day.replace("weekday_is_", "").capitalize()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x="Day", y="shares", data=df)
plt.title("Boxplot of Shares across Days of the Week")
plt.xticks(rotation=45)
plt.ylim(0,10000)   # limit to reduce outlier effect
plt.show()
```

```python
#Shares across Weekend vs Weekday
df['Weekend'] = df[' is_weekend'].map({0: "Weekday", 1: "Weekend"})

plt.figure(figsize=(6,6))
sns.boxplot(x="Weekend", y="shares", data=df)
plt.title("Boxplot of Shares: Weekend vs Weekday")
plt.ylim(0,10000)
plt.show()
```

```python
# Boxplot: Shares across Article Categories (data channels)
channels = [" data_channel_is_lifestyle", " data_channel_is_entertainment",
            " data_channel_is_bus", " data_channel_is_socmed",
            " data_channel_is_tech", " data_channel_is_world"]

# Create a new column for channel name
for ch in channels:
    df.loc[df[ch] == 1, 'Channel'] = ch.replace("data_channel_is_", "").capitalize()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x="Channel", y="shares", data=df)
plt.title("Boxplot of Shares across Article Categories")
plt.xticks(rotation=45)
plt.ylim(0,10000)
plt.show()
```
