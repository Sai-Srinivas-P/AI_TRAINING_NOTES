# Titanic dataset to illustrate the three major types of variables:

## Unordered categorical

## Ordered categorical

## Quantitative/numeric

```python
#Load Titanic Dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset from seaborn
df = sns.load_dataset('titanic')
print(df.head())
```

```python
# Unordered Categorical Variables
# These are categories with no inherent order.
# Identify unordered categorical variables
unordered_cats = ['sex', 'embarked', 'class', 'who', 'deck', 'embark_town']

# Display value counts for each
for col in unordered_cats:
    print(f"\nUnordered Categorical Variable: {col}")
    print(df[col].value_counts())

    # Plot distribution
    plt.figure(figsize=(6, 3))
    sns.countplot(x=col, data=df, order=df[col].value_counts().index)
    plt.title(f'Distribution of {col}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Ordered Categorical Variables
# These have a logical order, even if not numeric.
# Create an ordered categorical variable: Age Group
bins = [0, 12, 18, 60, 100]
labels = ['Child', 'Teen', 'Adult', 'Senior']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, ordered=True)

# Confirm it's ordered
print("\nOrdered Categorical Variable: age_group")
print(df['age_group'].value_counts().sort_index())

# Plot
plt.figure(figsize=(6, 3))
sns.countplot(x='age_group', data=df, order=labels)
plt.title('Passenger Count by Age Group')
plt.tight_layout()
plt.show()
```

```python
#Quantitative / Numeric Variables
# These are continuous or discrete numeric values.
# Identify numeric variables
numeric_vars = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
print("\nQuantitative Variables:", numeric_vars)

# Summary and distribution plots
for col in numeric_vars:
    print(f"\n Summary of {col}")
    print(df[col].describe())

    plt.figure(figsize=(6, 3))
    sns.histplot(df[col].dropna(), kde=True)
    plt.title(f'Distribution of {col}')
    plt.tight_layout()
    plt.show()
```
