```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Sample dataset
df = pd.DataFrame({
    "Age": [10, 15, 25, 35, 50, 70],
    "Math_Score": [45, 60, 75, 80, 90, 55],
    "English_Score": [40, 65, 70, 78, 85, 50],
    "Units_Sold": [10, 20, 15, 30, 40, 50],
    "Price": [100, 200, 150, 300, 250, 400],
    "Cost": [60, 120, 80, 200, 180, 300],
    "Weight": [50, 65, 70, 80, 90, 100],
    "Height": [1.5, 1.7, 1.8, 1.75, 1.6, 1.85]
})
print("Sample Dataset:\n", df.head(), "\n")
```

```python
#Derived metrics = new variables created from existing ones.
```

```python
#Type-Driven Metrics
```

```python
def age_group(age):
    if age < 13: return "Child"
    elif age < 20: return "Teen"
    elif age < 60: return "Adult"
    else: return "Senior"

df["Age_Group"] = df["Age"].apply(age_group)
print(df[["Age", "Age_Group"]])

plt.figure(figsize=(6,3))
sns.countplot(x="Age_Group", data=df, order=["Child", "Teen", "Adult", "Senior"])
plt.title("Type-Driven Metric: Age Groups")
plt.show()
```

```python
#Business-Driven Metrics
```

```python
df["Revenue"] = df["Units_Sold"] * df["Price"]
df["Profit_Margin"] = (df["Revenue"] - df["Cost"]) / df["Revenue"]

print(df[["Units_Sold", "Price", "Cost", "Revenue", "Profit_Margin"]])

plt.figure(figsize=(6,3))
sns.barplot(x="Units_Sold", y="Profit_Margin", data=df, color="orange")
plt.title("Business-Driven Metric: Profit Margin")
plt.show()
```

```python
#Data-Driven Metrics
```

```python
# BMI = Weight / Height²
df["BMI"] = df["Weight"] / (df["Height"]**2)

# Price per unit (derived metric)
df["Price_per_Unit"] = df["Revenue"] / df["Units_Sold"]

print(df[["Weight", "Height", "BMI", "Price_per_Unit"]])

plt.figure(figsize=(6,3))
sns.histplot(df["BMI"], bins=5, kde=True, color="green")
plt.title("Data-Driven Metric: BMI Distribution")
plt.show()
```

```python
print("Measurement Scales (Nominal, Ordinal, Interval, Ratio)")
print("Nominal: Categories (Zip code)")
print("Ordinal: Ordered categories (Education level)")
print("Interval: Differences meaningful, no true zero (°C)")
print("Ratio: Differences + true zero (Age, Income)\n")

#Nominal - gender, zip code, Names, 
#Ordinal - Order - not in equal intervals, - Rank, customer survey
#Inerval - order, in equal intervals - Dates, temperature
#Ratio - age, weight, salary
```

```python
#Discrete vs Continuous
#Discrete = Countable (Units_Sold)
#Continuous = Measurable (Height, Weight)
```
