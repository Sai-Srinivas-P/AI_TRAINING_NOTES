```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
```

```python
#Create a Sample Dataset
np.random.seed(42)
```

```python
# Generate study hours (continuous variable)
study_hours = np.random.randint(1, 11, 50)  # values between 1 and 10
```

```python
# Generate scores with some correlation to study_hours
math_scores = study_hours * 8 + np.random.randint(0, 15, 50)   # linear relation with noise
reading_scores = study_hours * 7 + np.random.randint(0, 20, 50)
```

```python
# Put into a DataFrame
df = pd.DataFrame({
    "Study_Hours": study_hours,
    "Math_Score": math_scores,
    "Reading_Score": reading_scores
})
```

```python
print("First 5 rows of dataset:")
print(df.head())
```

```python
#2. Scatterplot - Study Hours vs Math Score
plt.figure(figsize=(7,5))
sns.scatterplot(x="Study_Hours", y="Math_Score", data=df, color="blue", s=70)
plt.title("Study Hours vs Math Score")
plt.xlabel("Study Hours")
plt.ylabel("Math Score")
plt.show()
```

```python
#3. Correlation (Study Hours vs Math Score)
corr, p_value = pearsonr(df["Study_Hours"], df["Math_Score"])
print(f"Correlation between Study Hours & Math Score = {corr:.2f}, p-value = {p_value:.4f}")
```

```python
#4. Scatterplot - Math Score vs Reading Score (Continuous vs Continuous)
plt.figure(figsize=(7,5))
sns.regplot(x="Math_Score", y="Reading_Score", data=df, color="green", ci=None)
plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

```python
# Correlation (Math vs Reading)
corr2, p_value2 = pearsonr(df["Math_Score"], df["Reading_Score"])
print(f"Correlation between Math & Reading Scores = {corr2:.2f}, p-value = {p_value2:.4f}")
```
