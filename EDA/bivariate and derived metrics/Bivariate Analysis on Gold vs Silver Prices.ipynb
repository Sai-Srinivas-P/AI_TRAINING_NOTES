```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
```

```python
df = pd.read_csv("D:\\Phani\\etechguild\\syllabus\\demo\\lessons\\python\\EDA\\data sheets\\EDA_Gold_Silver_prices.csv")
```

```python
# Convert Month column to datetime
df["Month"] = pd.to_datetime(df["Month"], format="%b-%y", errors="coerce")

# Extract Year for segmentation
df["Year"] = df["Month"].dt.year

print("Dataset Preview:")
print(df.head())
```

```python
#Scatterplot: Silver vs Gold
plt.figure(figsize=(7,5))
sns.scatterplot(x="SilverPrice", y="GoldPrice", data=df, s=70, color="purple")
plt.title("Silver Price vs Gold Price")
plt.xlabel("Silver Price")
plt.ylabel("Gold Price")
plt.show()
```

```python
# Correlation
corr, p_value = pearsonr(df["SilverPrice"], df["GoldPrice"])
print(f"Overall Correlation between Silver & Gold Prices = {corr:.2f}, p-value = {p_value:.4f}")
```

```python
# Regression Plot
plt.figure(figsize=(7,5))
sns.regplot(x="SilverPrice", y="GoldPrice", data=df, ci=None, color="orange")
plt.title("Gold vs Silver Price with Regression Line")
plt.xlabel("Silver Price")
plt.ylabel("Gold Price")
plt.show()
```

```python
# Time Series Trend
plt.figure(figsize=(10,5))
plt.plot(df["Month"], df["SilverPrice"], marker="o", label="Silver Price")
plt.plot(df["Month"], df["GoldPrice"], marker="s", label="Gold Price")
plt.title("Gold & Silver Prices Over Time")
plt.xlabel("Month")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()
```
