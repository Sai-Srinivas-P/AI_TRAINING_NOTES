```python
# Bivariate Analysis on Gold & Silver Prices
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
```

```python
data = {
    "Month": ["Mar-97","Apr-97","May-97","Jun-97","Jul-97","Aug-97","Sep-97",
              "Oct-97","Nov-97","Dec-97","Jan-98","Feb-98","Mar-98","Apr-98"],
    "SilverPrice": [186.48,170.65,170.44,169.96,155.8,162.16,172.13,
                    181.57,187.99,224.16,230.08,261.75,243.81,249.37],
    "GoldPrice": [12619.24,12338.59,12314.94,12202.78,11582.07,11640.38,
                  11755.57,11769.89,11370.19,11324.79,11378.25,11576.23,
                  11690.84,12225.02]
}

df = pd.DataFrame(data)
```

```python
# Convert Month to datetime for time-series plotting
df["Month"] = pd.to_datetime(df["Month"], format="%b-%y")
```

```python
print("Dataset Preview:")
print(df.head())
```

```python
#Scatterplot: Silver vs Gold
plt.figure(figsize=(7,5))
sns.scatterplot(x="SilverPrice", y="GoldPrice", data=df, s=70, color="purple")
plt.title("Silver Price vs Gold Price")
plt.xlabel("Silver Price")
plt.ylabel("Gold Price")
plt.show()
```

```python
#Correlation
corr, p_value = pearsonr(df["SilverPrice"], df["GoldPrice"])
print(f"Correlation between Silver & Gold Prices = {corr:.2f}, p-value = {p_value:.4f}")
```

```python
#Regression Plot
plt.figure(figsize=(7,5))
sns.regplot(x="SilverPrice", y="GoldPrice", data=df, ci=None, color="orange")
plt.title("Gold vs Silver Price with Regression Line")
plt.xlabel("Silver Price")
plt.ylabel("Gold Price")
plt.show()
```

```python
#Time Series Trend
plt.figure(figsize=(10,5))
plt.plot(df["Month"], df["SilverPrice"], marker="o", label="Silver Price")
plt.plot(df["Month"], df["GoldPrice"], marker="s", label="Gold Price")
plt.title("Gold & Silver Prices Over Time")
plt.xlabel("Month")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()
```

Scatterplot → See the relationship between Silver & Gold prices.

Correlation → Numeric measure of their relationship.

Regression line → Trend line between the two.

Time-series line plot → How Gold & Silver prices move across months.
