```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, ttest_ind, chi2_contingency
```

```python
df = pd.read_csv("D:\\Phani\\etechguild\\syllabus\\demo\\lessons\\python\\EDA\\data sheets\\Titanic-Dataset.csv")
```

```python
# Keep only relevant columns
df = df[['Age', 'Fare', 'Sex', 'Survived', 'Pclass']]
```

```python
# Drop missing values for simplicity
df = df.dropna()
```

```python
# Continuous vs Continuous
print("\n🔹 Continuous vs Continuous (Age vs Fare)")
corr, pval = pearsonr(df['Age'], df['Fare'])
print(f"Pearson Correlation: {corr:.2f}, p-value: {pval:.4f}")
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='Age', y='Fare', data=df, alpha=0.7)
sns.regplot(x='Age', y='Fare', data=df, scatter=False, color='red')
#best fit linear regression line in red, shows the trend between age and fare (whether higher age passengers paid more or less)
plt.title("Age vs Fare (Continuous vs Continuous)")
plt.show()
```

```python
#Categorical vs Continuous
print("\n🔹 Categorical vs Continuous (Sex vs Age)")
print(df.groupby("Sex")['Age'].mean())

male_age = df[df['Sex']=='male']['Age']
female_age = df[df['Sex']=='female']['Age']
t_stat, p_val = ttest_ind(male_age, female_age)
print(f"T-test: statistic={t_stat:.2f}, p-value={p_val:.4f}")
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='Sex', y='Age', data=df)
plt.title("Age Distribution by Sex (Categorical vs Continuous)")
plt.show()
```

```python
#Categorical vs Categorical
print("\n🔹 Categorical vs Categorical (Sex vs Survived)")
ct = pd.crosstab(df['Sex'], df['Survived'])
print(ct)

chi2, p, dof, expected = chi2_contingency(ct)
print(f"Chi-square Test: chi2={chi2:.2f}, p-value={p:.4f}")
```

```python
plt.figure(figsize=(6,5))
sns.heatmap(ct, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Sex vs Survival (Categorical vs Categorical)")
plt.show()
```
