```python
# 📌 Step 0: Load the Dataset
import pandas as pd

# Load Titanic dataset (you can replace with your own path or use seaborn's version)
df = pd.read_csv('D:\\Phani\\etechguild\\syllabus\\demo\\lessons\\python\\EDA\\Titanic-Dataset.csv') 
df.head()
```

```python
#checklist
# Checklist: Track your progress
cleaning_steps = {
    "Fix Rows": False,
    "Fix Columns": False,
    "Fix Missing Values": False,
    "Standardize Values": False,
    "Fix Invalid Values": False,
    "Filter Data": False
}
```

```python
#fix rows
# Remove irrelevant rows (if any)
# Titanic dataset is clean in terms of summary/footer rows, but here's how you'd do it:
df = df[df['Name'].notna()]  # Example: Remove rows where 'Name' is missing

# Reset index after row removal
df.reset_index(drop=True, inplace=True)
```

```python
#fix columns
# Rename columns for clarity
df.rename(columns={'Pclass': 'Passenger_Class', 'Fare': 'Ticket_Fare'}, inplace=True)

# Merge columns: Create Full Name
df['Full_Name'] = df['Name']

# Split column: Extract Title from Name
df['Title'] = df['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)

# Drop unnecessary columns
df.drop(columns=['Cabin'], inplace=True)  # Cabin has too many missing values
```

```python
#fix missing values
# Check missing values
df.isnull().sum()

# Fill missing Age with median
df.fillna({'Age': df['Age'].median()}, inplace=True)


# Fill missing Embarked with mode
df.fillna({'Embarked': df['Embarked'].mode()[0]}, inplace=True)
```

```python
# Standardize gender values
df['Sex'] = df['Sex'].str.lower()

# Standardize Embarked values
df['Embarked'] = df['Embarked'].replace({'S': 'Southampton', 'C': 'Cherbourg', 'Q': 'Queenstown'})
```

```python
# Remove invalid ages (e.g., negative values)
df = df[df['Age'] >= 0]

# Filter passengers with fare > 0
df = df[df['Ticket_Fare'] > 0]

# Filter dataset to only adult passengers
df_adults = df[df['Age'] >= 18]
```
