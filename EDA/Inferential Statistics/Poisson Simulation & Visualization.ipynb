```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import poisson
```

```python
# Simulation Function
def simulate_poisson(lambda_val, size=1000):
    return np.random.poisson(lam=lambda_val, size=size)
```

```python
# Visualization Function
def plot_poisson_distribution(lambda_vals, size=1000):
    plt.figure(figsize=(12, 6))
    for lam in lambda_vals:
        data = simulate_poisson(lam, size)
        sns.histplot(data, kde=False, stat="probability", bins=range(0, max(data)+2), label=f"λ={lam}", alpha=0.6)
    plt.title("Poisson Distribution for Different λ Values")
    plt.xlabel("Number of Events")
    plt.ylabel("Probability")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
lambda_values = [1, 4, 10]
plot_poisson_distribution(lambda_values)
```
