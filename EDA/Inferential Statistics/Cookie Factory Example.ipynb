```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# -----------------------------
# Cookie Factory Parameters
# -----------------------------
mu = 100            # Claimed mean weight (H₀)
sample_mean = 98.5  # Observed sample mean
sigma = 2           # Known population standard deviation
n = 30              # Sample size
alpha = 0.05        # Significance level


```

```python
# -----------------------------
#  Z-score Calculation
# -----------------------------
z = (sample_mean - mu) / (sigma / np.sqrt(n))
    p_value = stats.norm.cdf(z)  # One-tailed test (left side)

```

```python
# -----------------------------
# Decision
# -----------------------------
print(f"Z-score: {z:.4f}")
print(f"P-value: {p_value:.4f}")
if p_value < alpha:
    print("Reject H₀: The machine is likely underfilling.")
else:
    print("Fail to reject H₀: No strong evidence of underfilling.")

```

```python
# -----------------------------
# Visualization
# -----------------------------
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
y = stats.norm.pdf(x, mu, sigma / np.sqrt(n))

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Sampling Distribution (H₀)', color='blue')
plt.fill_between(x, y, where=(x < stats.norm.ppf(alpha, mu, sigma / np.sqrt(n))), color='red', alpha=0.3, label='Rejection Region')
plt.axvline(sample_mean, color='green', linestyle='--', label=f'Sample Mean = {sample_mean}g')
plt.title('One-Sample Z-Test: Cookie Weight')
plt.xlabel('Cookie Weight (grams)')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```
