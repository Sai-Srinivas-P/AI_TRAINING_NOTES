```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import nbinom
```

```python
def simulate_negative_binomial(r, p, size=1000):
    return np.random.negative_binomial(n=r, p=p, size=size)
```

```python
def plot_negative_binomial_distribution(r_values, p_values, size=1000):
    plt.figure(figsize=(12, 6))
    for r, p in zip(r_values, p_values):
        data = simulate_negative_binomial(r, p, size)
        sns.histplot(data, kde=False, stat="probability", bins=range(0, max(data)+2), label=f"r={r}, p={p}", alpha=0.6)
    plt.title("Negative Binomial Distribution for Different r and p Values")
    plt.xlabel("Trial Number of r-th Success")
    plt.ylabel("Probability")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
r_values = [2, 5, 10]
p_values = [0.2, 0.5, 0.7]
plot_negative_binomial_distribution(r_values, p_values)
```
