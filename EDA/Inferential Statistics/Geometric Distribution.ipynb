```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import geom
```

```python
#Simulation Function
def simulate_geometric(p, size=1000):
    return np.random.geometric(p, size=size)
```

```python
#Visualization Function
def plot_geometric_distribution(p_values, size=1000):
    plt.figure(figsize=(12, 6))
    for p in p_values:
        data = simulate_geometric(p, size)
        sns.histplot(data, kde=False, stat="probability", bins=range(1, max(data)+2), label=f"p={p}", alpha=0.6)
    plt.title("Geometric Distribution for Different p Values")
    plt.xlabel("Trial Number of First Success")
    plt.ylabel("Probability")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
p_values = [0.1, 0.3, 0.7]
plot_geometric_distribution(p_values)
```
