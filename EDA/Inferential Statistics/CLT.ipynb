```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Parameters
num_samples = 10000   # how many times we repeat the experiment
sample_size = 30      # number of die rolls in each experiment
```

```python
# Store sample means
sample_means = []
```

```python
# Run experiments
for _ in range(num_samples):
    rolls = np.random.randint(1, 7, sample_size)  # roll a die (1 to 6)
    sample_means.append(np.mean(rolls))           # take the mean
```

```python
# Plot the histogram of sample means
plt.figure(figsize=(8,5))
plt.hist(sample_means, bins=50, density=True, color="skyblue", edgecolor="black")
plt.title(f"Central Limit Theorem - Die Roll Averages (n={sample_size})")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```
