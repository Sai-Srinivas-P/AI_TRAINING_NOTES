```python
#Line Plot
import matplotlib.pyplot as plt

x = [1, 2, 3, 4,5] # iterations etc.,
y = [10, 20, 25, 30,100] # accuracy etc.,

plt.plot(x, y, marker='o', label='Line')
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Scatter Plot
x = [5, 7, 8, 7]
y = [99, 86, 87, 88]

plt.scatter(x, y, color='red')
plt.title('Scatter Plot')
plt.xlabel('Feature A')
plt.ylabel('Feature B')
plt.grid(True)
plt.show()

#feature correlation - visualize how two features relate 9ex: height vs weight)
```

```python
#Vertical Bar Chart
categories = ['A', 'B', 'C']
values = [10, 20, 15]

plt.bar(categories, values, color='skyblue')
plt.title('Bar Chart')
plt.ylabel('Values')
plt.show()

#show average values per category (ex: mean income by region)
```

```python
#Horizontal Bar Chart
plt.barh(categories, values, color='green')
plt.title('Horizontal Bar Chart')
plt.xlabel('Values')
plt.show()
```

```python
#Histogram
data = [22, 87, 5, 43, 56, 73, 55, 54, 11, 20, 51, 5, 79, 31, 27]
#raw numerical data (ex: scores, ages...)
plt.hist(data, bins=5, color='orange', edgecolor='black')
plt.title('Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

# distribution of numerical data - used in preprocessing, normalization and outlier detection
```

```python
# Box Plot
data = [7, 15, 13, 18, 21, 22, 23, 24, 25, 30, 35]
plt.boxplot(data)
plt.title('Box Plot')
plt.show()
```

```python
#Pie Chart
labels = ['Python', 'Java', 'C++']
sizes = [215, 130, 245]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Pie Chart')
plt.axis('equal')  # Equal aspect ratio ensures pie is circular
plt.show()
```

```python
#Customization Methods
plt.plot([1, 2, 3], [4, 5, 6], label='Line')
plt.title('Custom Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Axis Limits
plt.plot([1, 2, 3], [4, 5, 6])
plt.xlim(0, 4)
plt.ylim(3, 7)
plt.show()
```

```python
# Custom Tick Labels
plt.plot([1, 2, 3], [4, 5, 6])
plt.xticks([1, 2, 3], ['One', 'Two', 'Three'])
plt.yticks([4, 5, 6], ['Low', 'Medium', 'High'])
plt.show()
```

```python
#Figure & Subplot Layout
plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3], [4, 5, 6])
plt.title('New Figure')
plt.show()
```

```python
# Grid of Plots (Manual)
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [4, 5, 6])
plt.title('Left Plot')

plt.subplot(1, 2, 2)
plt.plot([1, 2, 3], [6, 5, 4])
plt.title('Right Plot')

plt.tight_layout()
plt.show()
```

```python
# Grid with Axes Objects
fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].bar(['A', 'B'], [10, 20])
axs[0].set_title('Bar Chart')

axs[1].scatter([1, 2], [3, 4])
axs[1].set_title('Scatter Plot')

plt.tight_layout()
plt.show()
```

```python
# Saving Plots
plt.plot([1, 2, 3], [4, 5, 6])
plt.title('Save This Plot')
plt.savefig('my_plot.png')
```

```python
#Comparing Multiple Features - box plot
data = [
    [10, 12, 14, 15, 18, 20, 22],  # Feature A
    [5, 7, 9, 10, 11, 13, 15],     # Feature B
    [20, 25, 30, 35, 40, 45, 50]   # Feature C
]

plt.boxplot(data, labels=['A', 'B', 'C'])
plt.title('Feature Distribution Comparison')
plt.show()
```
