```python
# 📦 Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# 📥 Load Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# 📌 Basic inspection
df.head()
```

```python
#1. Bar Chart: Passenger Count by Class
class_counts = df['Pclass'].value_counts().sort_index()

plt.bar(class_counts.index, class_counts.values, color='skyblue')
plt.title('Passenger Count by Class')
plt.xlabel('Class')
plt.ylabel('Count')
plt.grid(True)
plt.show()
```

```python
#Line Plot: Survival Rate by Age Group
df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 12, 18, 30, 50, 80], labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'])
survival_rate = df.groupby('AgeGroup')['Survived'].mean()

plt.plot(survival_rate.index, survival_rate.values, marker='o', color='green')
plt.title('Survival Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Survival Rate')
plt.grid(True)
plt.show()
```

```python
#3. Histogram: Age Distribution
plt.hist(df['Age'].dropna(), bins=20, color='orange', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
#4. Box Plot: Fare by Class
df.boxplot(column='Fare', by='Pclass', grid=False)
plt.title('Fare Distribution by Class')
plt.suptitle('')
plt.xlabel('Class')
plt.ylabel('Fare')
plt.show()
```

```python
#5. Scatter Plot: Age vs. Fare
plt.scatter(df['Age'], df['Fare'], alpha=0.5, color='purple')
plt.title('Age vs. Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True)
plt.show()
```

```python
#6. Subplots: Survival by Gender and Class
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Survival by Gender
gender_survival = df.groupby('Sex')['Survived'].mean()
axs[0].bar(gender_survival.index, gender_survival.values, color='teal')
axs[0].set_title('Survival Rate by Gender')
axs[0].set_ylabel('Survival Rate')

# Survival by Class
class_survival = df.groupby('Pclass')['Survived'].mean()
axs[1].bar(class_survival.index, class_survival.values, color='coral')
axs[1].set_title('Survival Rate by Class')
axs[1].set_xlabel('Class')

plt.tight_layout()
plt.show()

#HW - try survival rate by class by gender
```

```python
#7. Save a Plot
plt.plot(class_survival.index, class_survival.values, marker='o')
plt.title('Survival Rate by Class')
plt.xlabel('Class')
plt.ylabel('Survival Rate')
plt.grid(True)
plt.savefig('survival_by_class.png')
```
