# NumPy Advanced Operations

## 1. Multidimensional Arrays

```python
import numpy as np

# Example: Daily sales data for a week (3 products)
sales_data = np.array([[200, 220, 250], [180, 190, 210], [230, 240, 260], [210, 215, 225],
                       [250, 265, 275], [195, 205, 215], [210, 220, 230]])
print("Sales Data Shape:", sales_data.shape)
print("Sales Data:", sales_data)
```

## 2. Mathematical Operations on NumPy Arrays

```python
# Example: Increase all sales by 10%
increased_sales = sales_data * 1.10
print("Increased Sales:", increased_sales)
```

## 3. Reshape Arrays

```python
# Reshape to (3, 7)
reshaped_data = sales_data.T
print("Reshaped Data:", reshaped_data)
```

## 4. Stack Arrays

```python
# Stack new week's data
week2 = np.array([[210, 230, 255], [190, 200, 215], [240, 250, 270], [215, 220, 235],
                  [260, 275, 285], [200, 210, 225], [220, 230, 240]])

#to combine sales_data and week2 , 2 represents number of layers
# 7 is no.of rows(#days)
#3 no.of columns
stacked = np.stack((sales_data, week2))
print("Stacked Shape:", stacked.shape)

# organize multiple dataseets for comparison or input to models, combine data samples
```

```python
## 5. Basic Mathematical Operations
arr = np.array([-3, -2, 0, 2, 3])
print("Power:", np.power(arr, 2))
print("Absolute:", np.abs(arr))
print("Sine:", np.sin(arr))
print("Exponential:", np.exp(arr))
print("Logarithm (natural):", np.log(np.abs(arr) + 1))
```

```python
## 6. Apply Built-in Functions
print("Mean Sales:", np.mean(sales_data))
print("Max Sales per Day:", np.max(sales_data, axis=1))
print("Min Sales per Product:", np.min(sales_data, axis=0))
```

```python
## 8. Basic Linear Algebra Operations
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:", A)
print("Matrix B:", B)
print("Matrix Multiplication:", np.dot(A, B))
print("Transpose of A:", A.T)
print("Inverse of A:", np.linalg.inv(A))
#dot product - deep learning - 
#Transpose - reshaping data, optimize, aligning dimensions
#np.linalg.inv(A) - finding inverse of a matrix - linear regression, solving linear equations
```

```python
#Cosine similarity uses dot product in NLP
from numpy.linalg import norm
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
#cos_sim = dot(a,b) / (IIaII * IIbII)
cos_sim = np.dot(a, b) / (norm(a) * norm(b))
print("Cosine Similarity:", cos_sim)
```

```python
#Probability
data = np.array([1, 2, 3, 4, 5])

mean = np.mean(data)
variance = np.var(data)
std_dev = np.std(data)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std_dev)

#mean: average , tendency towards where the data centers ML --> predict average outcomes
#variance: how far the numbers are from the mean   ML --> understand uncertainity, noise levels in modeling
#standard deviation: square root of variance, gives the spread of the data ML --> understand uncertainity
```
