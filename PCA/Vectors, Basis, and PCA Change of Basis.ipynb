```python
# Step 1: Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Step 2: Define original 2D dataset (5 points)
data = np.array([[2, 3],
                 [3, 5],
                 [4, 2],
                 [5, 4],
                 [6, 5]])

# Step 3: Plot original vectors and standard basis
origin = np.zeros((2,))  # origin point

plt.figure(figsize=(8,8))
plt.scatter(data[:,0], data[:,1], color='blue', label='Data Points')

# Draw vectors from origin
for i in range(data.shape[0]):
    plt.arrow(origin[0], origin[1], data[i,0], data[i,1], 
              head_width=0.2, head_length=0.3, fc='blue', ec='blue')

# Standard basis vectors
i = np.array([1,0])
j = np.array([0,1])
plt.arrow(0,0,i[0],i[1], head_width=0.2, head_length=0.3, fc='red', ec='red')
plt.arrow(0,0,j[0],j[1], head_width=0.2, head_length=0.3, fc='green', ec='green')
plt.text(1.1, 0, 'i', color='red', fontsize=12)
plt.text(0, 1.1, 'j', color='green', fontsize=12)

plt.xlim(-1,7)
plt.ylim(-1,7)
plt.grid(True)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Original Data Vectors and Standard Basis')
plt.legend()
plt.show()

# Step 4: Apply PCA to find new basis (principal components)
pca = PCA(n_components=2)
pca.fit(data)
data_pca = pca.transform(data)

# Get principal components (new basis vectors)
pc1 = pca.components_[0]
pc2 = pca.components_[1]

# Step 5: Plot new basis and transformed data
plt.figure(figsize=(8,8))
plt.scatter(data_pca[:,0], data_pca[:,1], color='orange', label='Transformed Points')

# Draw new basis vectors
plt.arrow(0,0,pc1[0]*5, pc1[1]*5, head_width=0.2, head_length=0.3, fc='purple', ec='purple')
plt.arrow(0,0,pc2[0]*5, pc2[1]*5, head_width=0.2, head_length=0.3, fc='brown', ec='brown')
plt.text(pc1[0]*5.2, pc1[1]*5.2, 'PC1', color='purple', fontsize=12)
plt.text(pc2[0]*5.2, pc2[1]*5.2, 'PC2', color='brown', fontsize=12)

plt.xlim(-7,7)
plt.ylim(-7,7)
plt.grid(True)
plt.xlabel('PC1-axis')
plt.ylabel('PC2-axis')
plt.title('Data in PCA (New Basis) Space')
plt.legend()
plt.show()

# Step 6: Explained variance
explained_variance = pca.explained_variance_ratio_
print("Explained Variance Ratio:", explained_variance)
```
