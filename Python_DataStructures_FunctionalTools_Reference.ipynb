# Python Foundations in ML, AI & GenAI
### Dictionaries, Lists, Sets, Tuples, Comprehensions, Functions, Map, Filter

## 📋 Lists

**Definition**: Ordered, mutable collection allowing duplicates.

**Syntax**:

```
my_list = [1, 2, 3, 'AI', 4.5]
```

**Use Case**: Storing model predictions or GenAI prompts.

## 📘 Dictionaries

**Definition**: Unordered, mutable collection of key-value pairs with unique keys.

**Syntax**:

```
my_dict = {'model': 'GPT-4', 'accuracy': 0.95}
```

**Use Case**: Storing model scores or API request payloads.

## 🧵 Tuples

**Definition**: Ordered, immutable collection.

**Syntax**:

```
my_tuple = ('input_shape', 28, 28)
```

**Use Case**: Store input shapes or return multiple values.

## 🧮 Sets

**Definition**: Unordered, mutable collection of unique elements.

**Syntax**:

```
my_set = {1, 2, 3, 2, 1}  # Output: {1, 2, 3}
```

**Use Case**: Find unique labels or deduplicate NLP tokens.

## 🔁 Comprehensions

**Definition**: Short-hand for loops to generate collections.

**Examples**:

```
# List comprehension
squares = [x**2 for x in range(5)]

# Dict comprehension
squares_dict = {x: x**2 for x in range(3)}

# Set comprehension
unique = {x**2 for x in [1, 2, 2, 3]}
```

**Use Case**: Efficiently generate or transform datasets/prompts.

## 🧠 Functions

**Definition**: Reusable block of code to perform a task.

**Syntax**:

```
def greet(name):
    return f"Hello, {name}" 
```

**Use Case**: Metrics computation, API formatting.

## 🔄 map()

**Definition**: Applies a function to each item in a sequence.

**Syntax**:

```
squared = list(map(lambda x: x**2, [1, 2, 3]))
```

**Use Case**: Format predictions or clean GenAI outputs.

## 🔍 filter()

**Definition**: Filters elements based on a condition.

**Syntax**:

```
even = list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4]))
```

**Use Case**: Filter high-confidence results or lengthy chatbot responses.
