```python
import numpy as np

# Input features (e.g., size and number of rooms)
X = np.array([[-0.32], [-0.66]])  # shape: (2, 1)

# True output (house price)
y = np.array([[-0.54]])  # shape: (1, 1)

# Initial weights and biases
W1 = np.array([[0.2, 0.15], [0.5, 0.6]])  # shape: (2, 2)
b1 = np.array([[0.1], [0.25]])           # shape: (2, 1)

W2 = np.array([[0.3, 0.2]])              # shape: (1, 2)
b2 = np.array([[0.4]])                   # shape: (1, 1)

# Activation functions
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(h):
    return h * (1 - h)

# Learning rate
eta = 0.2
```

```python
#Forward Pass
```

```python
# Hidden layer
Z1 = np.dot(W1, X) + b1
H1 = sigmoid(Z1)

# Output layer (linear activation)
Z2 = np.dot(W2, H1) + b2
H2 = Z2  # linear activation

print("Predicted output:", H2)
```

```python
#Loss Computation (MSE)
```

```python
# Mean Squared Error
loss = 0.5 * (y - H2)**2
print("Loss:", loss)
```

```python
#Backward Pass
```

```python
# Output layer gradients
dL_dH2 = -(y - H2)
dH2_dZ2 = 1  # linear activation
dZ2_dW2 = H1.T
dZ2_db2 = 1

dL_dW2 = dL_dH2 * dH2_dZ2 * dZ2_dW2
dL_db2 = dL_dH2 * dH2_dZ2 * dZ2_db2

# Hidden layer gradients
dZ2_dH1 = W2.T
dL_dH1 = dL_dH2 * dH2_dZ2 * dZ2_dH1
dH1_dZ1 = sigmoid_derivative(H1)
dZ1_dW1 = X.T
dZ1_db1 = 1

dL_dW1 = dL_dH1 * dH1_dZ1 @ dZ1_dW1
dL_db1 = dL_dH1 * dH1_dZ1 * dZ1_db1
```

```python
#Parameter Update
```

```python
W2 -= eta * dL_dW2
b2 -= eta * dL_db2

W1 -= eta * dL_dW1
b1 -= eta * dL_db1
```

```python
#Repeat for Multiple Epochs
```

```python
for epoch in range(1000):
    # Forward pass
    Z1 = np.dot(W1, X) + b1
    H1 = sigmoid(Z1)
    Z2 = np.dot(W2, H1) + b2
    H2 = Z2

    # Loss
    loss = 0.5 * (y - H2)**2

    # Backward pass
    dL_dH2 = -(y - H2)
    dL_dW2 = dL_dH2 * H1.T
    dL_db2 = dL_dH2

    dL_dH1 = dL_dH2 * W2.T
    dL_dW1 = (dL_dH1 * sigmoid_derivative(H1)) @ X.T
    dL_db1 = dL_dH1 * sigmoid_derivative(H1)

    # Update
    W2 -= eta * dL_dW2
    b2 -= eta * dL_db2
    W1 -= eta * dL_dW1
    b1 -= eta * dL_db1

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item():.4f}")
```
