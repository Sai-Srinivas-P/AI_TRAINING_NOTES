```python
#2 input features (e.g., house size and number of rooms)

#1 hidden layer with 2 neurons

#1 output neuron (predicting house price)

#Sigmoid activation in hidden layer, linear activation in output

#Mean Squared Error as the loss function
```

```python
import numpy as np

X = np.array([[0.5], [0.1]])  # Input features
y = np.array([[0.6]])         # True output

W1 = np.array([[0.1, 0.2], [0.3, 0.4]])  # Hidden layer weights
b1 = np.array([[0.01], [0.02]])         # Hidden layer biases

W2 = np.array([[0.5, 0.6]])             # Output layer weights
b2 = np.array([[0.03]])                 # Output layer bias

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(h):
    return h * (1 - h)

eta = 0.1  # Learning rate
```

```python
#forward pass
```

```python
Z1 = np.dot(W1, X) + b1
H1 = sigmoid(Z1)

Z2 = np.dot(W2, H1) + b2
H2 = Z2  # Linear activation
```

```python
#loss compputation
loss = 0.5 * (y - H2)**2
```

```python
#Backward Pass
# Output layer gradients
dL_dH2 = -(y - H2)
dL_dW2 = dL_dH2 * H1.T
dL_db2 = dL_dH2

# Hidden layer gradients
dL_dH1 = dL_dH2 * W2.T
dH1_dZ1 = sigmoid_derivative(H1)
dL_dW1 = (dL_dH1 * dH1_dZ1) @ X.T
dL_db1 = dL_dH1 * dH1_dZ1
```

```python
#Parameter Update
W2 -= eta * dL_dW2
b2 -= eta * dL_db2
W1 -= eta * dL_dW1
b1 -= eta * dL_db1
```

```python
#Repeat and Evaluate
Z1_new = np.dot(W1, X) + b1
H1_new = sigmoid(Z1_new)
Z2_new = np.dot(W2, H1_new) + b2
H2_new = Z2_new
new_loss = 0.5 * (y - H2_new)**2
```
