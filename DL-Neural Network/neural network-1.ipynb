```python
#Perceptron
import numpy as np

# Inputs (e.g., weather, mood, free time)
x = np.array([1, 0.5, 0.8])     # sunny, moderately good mood, enough time
w = np.array([0.6, 0.3, 0.2])   # importance of each factor
b = -0.4                        # bias (default tendency to stay home)

# Compute cumulative input
z = np.dot(w, x) + b

# Step function
output = 1 if z > 0 else 0

print(f"Weighted Sum (z) = {z:.2f}")
print("Decision:", "Go out " if output == 1 else "Stay home ")
```

```python
#Visualizing the Step Function
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(-2, 2, 100)
y_vals = np.where(x_vals > 0, 1, 0)

plt.plot(x_vals, y_vals)
plt.title("Step Function (Perceptron Activation)")
plt.xlabel("Weighted Sum (z)")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

```python
#Artificial Neuron
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def relu(z):
    return np.maximum(0, z)

z = np.linspace(-5, 5, 100)
plt.plot(z, sigmoid(z), label="Sigmoid")
plt.plot(z, relu(z), label="ReLU")
plt.title("Activation Functions")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Feedforward Flow (Single Neuron)
x = np.array([0.6, 0.9, 0.3])
w = np.array([0.4, 0.2, 0.8])
b = -0.1

# Feedforward calculation
z = np.dot(w, x) + b
a = sigmoid(z)

print(f"z = {z:.3f}, activation (sigmoid) = {a:.3f}")
```

```python
#Mini Neural Network Example (2-Layer Feedforward)
# Hidden layer (2 neurons)
X = np.array([0.5, 0.2])   # inputs

W1 = np.array([[0.4, 0.3],
               [0.6, 0.9]])   # 2x2 weights
b1 = np.array([0.1, -0.2])

# Output layer
W2 = np.array([0.5, 0.7])
b2 = 0.3

# Hidden layer activations
z1 = np.dot(W1, X) + b1
a1 = sigmoid(z1)

# Output layer activation
z2 = np.dot(W2, a1) + b2
output = sigmoid(z2)

print(f"Hidden activations: {a1}")
print(f"Output: {output:.3f}")
```

```python
#Parameters & Hyperparameters in Scikit-Learn
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load data
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Define hyperparameters
model = MLPClassifier(hidden_layer_sizes=(5,), activation='relu',
                      learning_rate_init=0.01, max_iter=300, random_state=42)

# hidden_layer_sizes=(5,) - 1 hidden layer with 5 neurons
#
# Train model (learns parameters)
model.fit(X_train, y_train)
#it uses backpropagation and gradient descent to adjust weights and bias

print("Training Accuracy:", model.score(X_train, y_train))
print("Number of parameters (weights + biases):", sum(p.size for p in model.coefs_) + sum(b.size for b in model.intercepts_))
```

# Student Placement Prediction (Single Neuron)

```python
import numpy as np

# --- Step 1: Define input features ---
# Each value is normalized (between 0 and 1)
# Example student data:
# x1 = CGPA, x2 = Communication Skills, x3 = Internship Experience
x = np.array([0.6, 0.9, 0.3])

# --- Step 2: Define weights (importance of each feature) ---
# These can be learned in training; here we assume example values
w = np.array([0.4, 0.2, 0.8])

# --- Step 3: Define bias ---
b = -0.1

# --- Step 4: Define sigmoid activation function ---
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# --- Step 5: Feedforward computation ---
z = np.dot(w, x) + b          # Weighted sum
a = sigmoid(z)                # Activation (output probability)

# --- Step 6: Display results ---
print(f"Weighted sum (z): {z:.3f}")
print(f"Predicted probability of placement: {a:.3f}")

# --- Step 7: Interpret result ---
if a >= 0.5:
    print("Student is likely to be PLACED.")
else:
    print("Student is likely NOT to be placed.")
```
