```python
import numpy as np

# Step 1: Define the dataset
# Each row is a data point: [x1, x2]
X = np.array([
    [0, 3],
    [5, 9],
    [-1, -2]
])

# Labels: +1 for positive class, -1 for negative class
y = np.array([1, 1, -1])

# Step 2: Convert to homogeneous coordinates by appending 1 to each input
X_hom = np.hstack((X, np.ones((X.shape[0], 1))))

# Step 3: Define candidate weight vectors (including bias as last element)
# each candidate is a weight vector W=[w1,w2,b]
candidates = [
    np.array([-1, 1, -2]),
    np.array([-1, 1, -5]),
    np.array([-1, 2, -1]),
    np.array([-1, 2, 2])
]


# Step 4: Check which candidates are valid separators
def is_valid_separator(w, X_hom, y):
    predictions = np.sign(X_hom @ w)
    return np.array_equal(predictions, y)

# Step 5: Evaluate each candidate
for i, w in enumerate(candidates, start=1):
    valid = is_valid_separator(w, X_hom, y)
    print(f"Candidate {i}: {'Valid' if valid else 'Invalid'} Separator")
```
