```python
#Load the Titanic Dataset
import seaborn as sns
import pandas as pd

# Load Titanic dataset
titanic = sns.load_dataset('titanic')
titanic.head()
#note: Seaborn includes the Titanic dataset by default:
```

```python
#Survival Count by Gender
import matplotlib.pyplot as plt

# Example: converting 'survived' column to string
titanic['survived_str'] = titanic['survived'].astype(str)

# Now use the string column for hue
sns.countplot(x='sex', hue='survived_str', data=titanic)

plt.title("Survival Count by Gender")
plt.show()
```

```python
#Age Distribution by Survival
sns.histplot(data=titanic, x='age', hue='survived', multiple='stack', bins=30)
plt.title("Age Distribution by Survival")
plt.show()
#Younger passengers had slightly better survival odds.
```

```python
#Boxplot of Fare by Class
sns.boxplot(x='class', y='fare', data=titanic)
plt.title("Fare Distribution by Passenger Class")
plt.show()
#First-class passengers paid significantly higher fares.
```

```python
#Heatmap of Survival by Class
survival_by_class = titanic.groupby(['pclass', 'survived']).size().unstack()
#groupby pclass, survived - groups the data by passeneger class (pclass) and survival status
#size - counts the no.of rows in each group
#unstack - converts the second grouping level (survived) into columns
sns.heatmap(survival_by_class, annot=True, fmt='d', cmap='YlGnBu')
plt.title("Survival Count by Passenger Class")
plt.show()
#First-class passengers had the highest survival rate.
#annot=True - adds the actual numbers inside each cell
#fmt='d' - format the annotation as integers
#cmap='YlGnBu' - color scheme from yellow to green to blue
```
