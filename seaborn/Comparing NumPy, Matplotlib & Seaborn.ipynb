```python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
# modify this path matching your local (where file located?)
url = "D:\\Phani\\etechguild\\syllabus\\demo\\lessons\\python\\session-5 data visualization\\topic-2 seaborn\\googleplaystore.xlsx"
df = pd.read_excel(url)

# Basic Cleaning
df = df[df['Rating'].notnull()]
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
df = df[df['Rating'].notnull()]
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')
df = df[df['Reviews'].notnull()]
df = df[df['Rating'] <= 5]
```

```python
#NumPy: Feature Scaling & Summary Stats
# Feature Scaling: Normalize Reviews
reviews = df['Reviews'].values
scaled_reviews = (reviews - np.mean(reviews)) / np.std(reviews)

# Summary Stats
print("Mean Rating:", np.mean(df['Rating']))
print("Std Dev of Reviews:", np.std(reviews))
```

```python
#Matplotlib: Histogram & Scatter Plot
# Histogram of Ratings
plt.figure(figsize=(6,4))
plt.hist(df['Rating'], bins=20, color='skyblue', edgecolor='black')
plt.title("Rating Distribution (Matplotlib)")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

# Scatter Plot: Rating vs Reviews
plt.figure(figsize=(6,4))
plt.scatter(df['Reviews'], df['Rating'], alpha=0.3, color='purple')
plt.title("Rating vs Reviews (Matplotlib)")
plt.xlabel("Reviews")
plt.ylabel("Rating")
plt.xscale('log') # converts x-axis to logarithmic scale
plt.grid(True)
plt.show()
```

```python
#Seaborn: KDE, Boxplot, Heatmap
# KDE Plot of Ratings
plt.figure(figsize=(6,4))
sns.histplot(df['Rating'], bins=20, kde=True, color='orange')
plt.title("Rating Distribution (Seaborn)")
plt.xlabel("Rating")
plt.ylabel("Density")
plt.show()

# Boxplot: Rating by Content Rating
plt.figure(figsize=(8,4))
sns.boxplot(x='Content Rating', y='Rating', data=df)
plt.title("Rating by Content Rating")
plt.xticks(rotation=45)
plt.show()

# Heatmap of Correlations
plt.figure(figsize=(6,4))
sns.heatmap(df[['Rating', 'Reviews']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
