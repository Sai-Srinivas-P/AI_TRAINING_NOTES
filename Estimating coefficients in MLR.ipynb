```python
import numpy as np

# Design matrix (intercept + x1 + x2)
X = np.array([[1,1,2],
              [1,2,0],
              [1,3,1]])

Y = np.array([[4],
              [3],
              [5]])

beta = np.linalg.inv(X.T @ X) @ X.T @ Y
print("Beta coefficients (intercept, beta1, beta2):")
print(beta)   # should print [[1.],[1.],[1.]]
```
